```python
"""
Build and save all regional graphs
"""


# =========================================================
# Create an external import asset
# =========================================================

def create_external_import_asset(
    asset_handle,
    target_node,
    capacity_mw,
    operating_cost,
    source_label,
    jurisdiction,
    nerc,
):
    """
    Create a dispatchable external import.

    The import is modeled like a nonrenewable generator:

        _class = Producer
        renewable = False
        rps_eligible = False

    Therefore, it is not counted as eligible renewable
    generation. It is treated like gas generation in the
    RPS calculation.
    """

    capacity_mw = float(capacity_mw)
    operating_cost = float(operating_cost)

    return {
        "id": asset_handle,

        "oris_code": 0,
        "egrid_id": f"IMPORT_{asset_handle.upper()}",

        "type": "generator",
        "fuel": "import",

        # Critical classification:
        # Treat the import as actual generation, not demand.
        "_class": "Producer",

        "profile": f"{target_node}:generator:",
        "region": target_node,

        # Use the receiving region's jurisdiction.
        "jurisdiction": jurisdiction,
        "nerc": nerc,

        "utility": source_label,

        "x": 0.0,
        "y": 0.0,

        # GOOD uses watts internally.
        "installed_capacity": capacity_mw * 1e6,
        "capacity_mw": capacity_mw,

        "hydro_size_threshold_mw": None,
        "capacity_factor": 1.0,

        "dispatchable": True,
        "combinable": True,

        # Critical RPS classification:
        # Same treatment as gas turbines.
        "renewable": False,
        "rps_eligible": False,

        # Fixed interface capacity.
        "extensible": False,

        "capex_capacity": 0.0,
        "capex_cost": 0.0,

        # Cost in $/J.
        "operating_cost": operating_cost,

        "heat_rate": 0.0,

        # Emissions are not modeled because the generation
        # occurs outside the receiving model region.
        "nox": 0.0,
        "so2": 0.0,
        "co2": 0.0,
        "ch4": 0.0,
        "n2o": 0.0,
        "pm": 0.0,
    }


# =========================================================
# Build and save regional graphs
# =========================================================

saved_graphs = {}

for region_name, cfg in REGION_CONFIGS.items():

    print("=" * 80)
    print(f"Building {region_name}")
    print("=" * 80)

    deep_reload(good)

    # --------------------------------------------------
    # Load source graph
    # --------------------------------------------------

    graph = good.graph.graph_from_json(
        cfg["source_graph"]
    )

    print(
        "Source graph:",
        cfg["source_graph"],
    )

    print(
        "Requested nodes:",
        cfg["nodes"],
    )

    # --------------------------------------------------
    # Check requested nodes
    # --------------------------------------------------

    missing_nodes = [
        node
        for node in cfg["nodes"]
        if node not in graph.nodes
    ]

    if missing_nodes:
        raise ValueError(
            f"{region_name} has missing nodes: "
            f"{missing_nodes}. "
            f"Check the source graph: "
            f"{cfg['source_graph']}"
        )

    # --------------------------------------------------
    # Create regional subgraph
    # --------------------------------------------------

    graph = good.graph.subgraph(
        graph,
        cfg["nodes"],
    )

    print("Subgraph nodes:")
    print(graph.nodes)

    # --------------------------------------------------
    # Aggregate existing assets
    # --------------------------------------------------

    deep_reload(good)

    clustering = copy.deepcopy(
        good.aggregate.default_clustering
    )

    clustering["resolution"] = (
        AGGREGATION_RESOLUTION
    )

    graph = good.aggregate.aggregate(
        graph,
        clustering=clustering,
    )

    # --------------------------------------------------
    # Add configured external imports
    #
    # This happens after aggregation so that the import
    # remains a separate asset.
    # --------------------------------------------------

    region_import_configs = (
        EXTERNAL_IMPORT_CONFIGS.get(
            region_name,
            [],
        )
    )

    for import_cfg in region_import_configs:

        target_node = import_cfg["target_node"]
        asset_handle = import_cfg["asset_handle"]

        # Confirm the target node remains in the
        # aggregated regional graph.
        if target_node not in graph.nodes:
            raise ValueError(
                f"Cannot add {asset_handle}. "
                f"Target node {target_node} is not "
                f"in the {region_name} graph."
            )

        # Get the target node's asset dictionary.
        target_assets = (
            graph._node[target_node]
            .setdefault("assets", {})
        )

        # Prevent an accidental duplicate asset.
        if asset_handle in target_assets:
            raise ValueError(
                f"Cannot add {asset_handle}. "
                f"An asset with this name already "
                f"exists at {target_node}."
            )

        # Create the import.
        import_asset = (
            create_external_import_asset(
                asset_handle=asset_handle,
                target_node=target_node,
                capacity_mw=import_cfg["capacity_mw"],
                operating_cost=import_cfg["operating_cost"],
                source_label=import_cfg["source_label"],
                jurisdiction=import_cfg["jurisdiction"],
                nerc=import_cfg["nerc"],
            )
        )

        # Add the import to the target node.
        target_assets[asset_handle] = (
            import_asset
        )

        import_cost_per_mwh = (
            import_asset["operating_cost"]
            * 3.6e9
        )

        print(
            f"Added external import:"
            f"\n  Region: {region_name}"
            f"\n  Node: {target_node}"
            f"\n  Asset: {asset_handle}"
            f"\n  Source: "
            f"{import_cfg['source_label']}"
            f"\n  Capacity: "
            f"{import_asset['capacity_mw']:.1f} MW"
            f"\n  Cost: "
            f"${import_cost_per_mwh:.2f}/MWh"
        )

    # --------------------------------------------------
    # Validate configured imports
    # --------------------------------------------------

    for import_cfg in region_import_configs:

        target_node = import_cfg["target_node"]
        asset_handle = import_cfg["asset_handle"]

        saved_import = (
            graph._node[target_node]
            ["assets"][asset_handle]
        )

        expected_capacity_w = (
            float(import_cfg["capacity_mw"])
            * 1e6
        )

        if saved_import["fuel"] != "import":
            raise ValueError(
                f"{asset_handle} has an incorrect "
                f"fuel classification."
            )

        if saved_import["_class"] != "Producer":
            raise ValueError(
                f"{asset_handle} is not a Producer."
            )

        if saved_import["region"] != target_node:
            raise ValueError(
                f"{asset_handle} has an incorrect "
                f"target region."
            )

        if (
            saved_import["installed_capacity"]
            != expected_capacity_w
        ):
            raise ValueError(
                f"{asset_handle} has an incorrect "
                f"installed capacity."
            )

        if saved_import["rps_eligible"]:
            raise ValueError(
                f"{asset_handle} cannot be "
                f"RPS eligible."
            )

        print(
            f"Validated import asset "
            f"{target_node}:{asset_handle}"
        )

    # --------------------------------------------------
    # Save regional graph
    # --------------------------------------------------

    deep_reload(good)

    good.graph.graph_to_json(
        graph,
        cfg["output_graph"],
    )

    saved_graphs[region_name] = graph

    print(
        "Saved to:",
        cfg["output_graph"],
    )

    print()
```

```python
import json
import os
import pandas as pd

output_path0 = "Examples/Nodes/"
output_path1 = "Examples/"

# node_path = os.path.join(output_path0, "New_York_IPM.json")
# node_path = os.path.join(output_path0, "PJM_IPM.json")
node_path = os.path.join(output_path0, "MISO_IPM.json")
policies_path = os.path.join(output_path1, "policies.json")

with open(node_path, "r") as f:
    node_data = json.load(f)

with open(policies_path, "r") as f:
    policies = json.load(f)

print("Top-level keys:")
print(node_data.keys())

print("\nNumber of graph nodes:", len(node_data.get("nodes", [])))
print("Number of graph edges:", len(node_data.get("edges", [])))
```

```python
for i, node in enumerate(node_data["nodes"]):
    print("\nNode index:", i)
    print("Node keys:", node.keys())

    # Print possible node name fields
    for name_key in ["id", "name", "handle", "region", "label"]:
        if name_key in node:
            print(f"{name_key}:", node[name_key])

    if "assets" in node:
        print("Number of assets:", len(node["assets"]))
        print("First asset keys:", list(node["assets"].keys())[:5])
```

```python
all_assets = {}

for node in node_data["nodes"]:

    node_id = (
        node.get("id")
        or node.get("name")
        or node.get("handle")
        or node.get("region")
        or "unknown_node"
    )

    if "assets" not in node:
        continue

    node_assets = node["assets"]

    if not isinstance(node_assets, dict):
        print(f"Skipping node {node_id} because assets is not a dictionary")
        continue

    for asset_key, asset in node_assets.items():
        all_assets[f"{node_id}::{asset_key}"] = asset

assets = all_assets

print("Total extracted assets:", len(assets))
```

```python
missing_rps = [
    key for key, asset in assets.items()
    if isinstance(asset, dict) and "rps_eligible" not in asset
]

non_dict_items = [
    (key, type(asset).__name__)
    for key, asset in assets.items()
    if not isinstance(asset, dict)
]

print("Assets missing rps_eligible:", len(missing_rps))
print(missing_rps[:20])

print("Non-dict items:", len(non_dict_items))
print(non_dict_items[:20])
```
