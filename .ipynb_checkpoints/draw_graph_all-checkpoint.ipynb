```python
import os
import glob
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import good
```

```python
def load_total_costs_from_results(results_dir, scenarios=None, scenario_label=None):
    records = []

    total_files = glob.glob(os.path.join(results_dir, "**", "*_total_cost.csv"), recursive=True)
    summary_files = glob.glob(os.path.join(results_dir, "**", "*_summary.csv"), recursive=True)
    files = sorted(total_files + summary_files)

    print(f"Found {len(files)} candidate files in {results_dir}")

    seen = set()

    for fp in files:
        try:
            temp = pd.read_csv(fp)
            if temp.empty:
                continue

            row = temp.iloc[0].to_dict()

            fname = os.path.basename(fp)
            folder = os.path.basename(os.path.dirname(fp))

            scenario_id = int(fname.split("_")[0][1:])

            # prefer total_cost over summary
            if scenario_id in seen and fp.endswith("_summary.csv"):
                continue

            cfg = {}
            cfg["scenario_id"] = scenario_id
            cfg["scenario_tag"] = row.get("scenario_tag", "")
            cfg["objective_value"] = float(row["objective_value"])
            cfg["file_path"] = fp

            base_dir_name = os.path.basename(os.path.normpath(results_dir)).lower()

            if "_slow_" in f"_{base_dir_name}_":
                cfg["adoption"] = "Slow"
            elif "_fast_" in f"_{base_dir_name}_":
                cfg["adoption"] = "Fast"
            elif "_mid_" in f"_{base_dir_name}_":
                cfg["adoption"] = "Mid"
            else:
                cfg["adoption"] = "Unknown"

            # NEW: Add scenario label from directory
            if scenario_label is None:
                # Extract label from directory name (e.g., "flex", "midnight")
                parts = base_dir_name.split("_")
                cfg["scenario_label"] = parts[-1] if len(parts) > 0 else "default"
            else:
                cfg["scenario_label"] = scenario_label

            # Extract v1g_share and v2g_share from folder name
            m_v1g = re.search(r"v1g(\d+)", folder, re.IGNORECASE)
            m_v2g = re.search(r"v2g(\d+)", folder, re.IGNORECASE)

            v1g_val = int(m_v1g.group(1)) if m_v1g else 0
            v2g_val = int(m_v2g.group(1)) if m_v2g else 0

            cfg["v1g_share"] = v1g_val / 100.0
            cfg["v2g_share"] = v2g_val / 100.0

            # Determine participation level (10% or 30%)
            total_participation = v1g_val + v2g_val
            if total_participation == 0:
                cfg["participation"] = "0%"
                cfg["group"] = "Base only"
            elif total_participation == 10:
                cfg["participation"] = "10%"
            elif total_participation == 30:
                cfg["participation"] = "30%"
            elif total_participation == 50:
                cfg["participation"] = "50%"
            else:
                cfg["participation"] = f"{total_participation}%"

            # Determine group
            if v1g_val > 0 and v2g_val > 0:
                cfg["group"] = "V1G+V2G"
            elif v1g_val > 0:
                cfg["group"] = "V1G"
            elif v2g_val > 0:
                cfg["group"] = "V2G"
            else:
                cfg["group"] = "Base only"

            m = re.search(r"rps(\d+)", folder)
            cfg["rps_ratio"] = int(m.group(1)) / 100 if m else None

            m = re.search(r"bcapex(\d+)", folder)
            if m:
                batt = int(m.group(1))
                cfg["batt_capex_label"] = f"${batt}/kWh"
            else:
                cfg["batt_capex_label"] = None

            records = [r for r in records if r["scenario_id"] != scenario_id]
            records.append(cfg)
            seen.add(scenario_id)

        except Exception as e:
            print(f"Could not read {fp}: {e}")

    df = pd.DataFrame(records)

    if df.empty:
        raise ValueError(f"No valid total cost or summary files found in: {results_dir}")

    df = df.sort_values("scenario_id").reset_index(drop=True)
    print("Available scenario IDs:", sorted(df["scenario_id"].unique().tolist()))
    print("Participation levels found:", df["participation"].dropna().unique())

    return df

def build_state_results_dirs(state):
    return {
        "flex": [
            f"scenario_results_2030_slow_even_{state}_flex",
            f"scenario_results_2030_mid_even_{state}_flex",
            f"scenario_results_2030_fast_even_{state}_flex",
        ],
        "midnight": [
            f"scenario_results_2030_slow_even_{state}_midnight",
            f"scenario_results_2030_mid_even_{state}_midnight",
            f"scenario_results_2030_fast_even_{state}_midnight",
        ],
        "arrive": [
            f"scenario_results_2030_slow_even_{state}_arrive",
            f"scenario_results_2030_mid_even_{state}_arrive",
            f"scenario_results_2030_fast_even_{state}_arrive",
        ],
    }


def load_one_state_total_costs(
    state,
    batt_capex=150,
    program_scenario_label="flex",
    baseline_scenario_label="flex",
):
    """
    Loads total cost data for one state.

    For Base only:
        Uses baseline_scenario_label.

    For V1G, V2G, V1G+V2G:
        Uses program_scenario_label.

    Uses Mid adoption only for the main points.
    """

    results_dirs_dict = build_state_results_dirs(state)

    df_list = []

    for scenario_label, dirs in results_dirs_dict.items():
        for rd in dirs:
            if not os.path.exists(rd):
                print(f"Skipping missing folder: {rd}")
                continue

            temp = load_total_costs_from_results(
                results_dir=rd,
                scenarios=None,
                scenario_label=scenario_label,
            )

            df_list.append(temp)

    if len(df_list) == 0:
        raise ValueError(f"No data found for {state}")

    df = pd.concat(df_list, ignore_index=True)

    df["state"] = state
    df["rps_percent"] = (df["rps_ratio"] * 100).round().astype(int)
    df["batt_capex_num"] = df["batt_capex_label"].str.extract(r"(\d+)").astype(float)
    df["objective_value_mil"] = df["objective_value"] / 1e6

    df = df[df["batt_capex_num"] == batt_capex].copy()
    df = df[df["adoption"] == "Mid"].copy()

    base = df[
        (df["group"] == "Base only") &
        (df["scenario_label"] == baseline_scenario_label)
    ].copy()

    programs = df[
        (df["group"].isin(["V1G", "V2G", "V1G+V2G"])) &
        (df["scenario_label"] == program_scenario_label)
    ].copy()

    out = pd.concat([base, programs], ignore_index=True)

    return out


def plot_rps_reached_at_same_cost(
    states=("CA", "NY", "TX"),
    state_names=None,
    batt_capex=150,
    program_scenario_by_state=None,
    baseline_scenario_by_state=None,
    rps_order_by_state=None,
    selected_participation=("30%", "50%"),
    save_path=None,
    log_x=False,
):
    """
    Final combined figure.

    Columns:
        One panel per state.

    x axis:
        Total system cost, million dollars.

    y axis:
        RPS target reached.

    Colors:
        Charging strategy.

    Markers:
        Participation level.

    Dashed vertical lines:
        Baseline costs at each RPS level for each state.

    log_x:
        If True, use logarithmic x axis.
        If False, use normal linear x axis.
    """

    from matplotlib.lines import Line2D

    if state_names is None:
        state_names = {
            "CA": "California",
            "NY": "New York",
            "TX": "Texas",
        }

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    all_data = []

    for state in states:
        temp = load_one_state_total_costs(
            state=state,
            batt_capex=batt_capex,
            program_scenario_label=program_scenario_by_state[state],
            baseline_scenario_label=baseline_scenario_by_state[state],
        )
        all_data.append(temp)

    df = pd.concat(all_data, ignore_index=True)

    df = pd.concat(
        [
            df[
                (df["state"] == state) &
                (df["rps_percent"].isin(rps_order_by_state[state]))
            ]
            for state in states
        ],
        ignore_index=True
    )

    df = df[
        (df["group"] == "Base only") |
        (df["participation"].isin(selected_participation))
    ].copy()

    group_order = ["Base only", "V1G", "V2G", "V1G+V2G"]

    group_colors = {
        "Base only": "black",
        "V1G": "#4E79A7",
        "V2G": "#59A14F",
        "V1G+V2G": "#E15759",
    }

    marker_map = {
        "0%": "o",
        "30%": "s",
        "50%": "^",
    }

    fig, axes = plt.subplots(
        1,
        len(states),
        figsize=(16, 5),
        sharey=False,
        facecolor="white",
    )

    if len(states) == 1:
        axes = [axes]

    for ax, state in zip(axes, states):
        state_rps_order = rps_order_by_state[state]
        sub_state = df[df["state"] == state].copy()

        ax.set_facecolor("white")

        # -----------------------------
        # Baseline reference lines
        # -----------------------------
        base_rows = sub_state[
            sub_state["group"] == "Base only"
        ].sort_values("rps_percent")

        for _, row in base_rows.iterrows():
            base_cost = row["objective_value_mil"]
            base_rps = row["rps_percent"]

            ax.axvline(
                base_cost,
                color="gray",
                linestyle="--",
                linewidth=1.2,
                alpha=0.75,
                zorder=1,
            )

            ax.text(
                base_cost,
                max(state_rps_order) + 1,
                f"Base\n{base_rps}%",
                ha="center",
                va="bottom",
                fontsize=12,
                color="dimgray",
            )

        # -----------------------------
        # Strategy lines
        # -----------------------------
        for group in group_order:
            sub_group = sub_state[sub_state["group"] == group].copy()

            if sub_group.empty:
                continue

            if group == "Base only":
                plot_data = sub_group.sort_values("rps_percent")

                ax.plot(
                    plot_data["objective_value_mil"],
                    plot_data["rps_percent"],
                    color=group_colors[group],
                    marker=marker_map["0%"],
                    linewidth=2.2,
                    markersize=8,
                    label="Base",
                    zorder=4,
                )

            else:
                for part in selected_participation:
                    plot_data = sub_group[
                        sub_group["participation"] == part
                    ].sort_values("rps_percent")

                    if plot_data.empty:
                        continue

                    ax.plot(
                        plot_data["objective_value_mil"],
                        plot_data["rps_percent"],
                        color=group_colors[group],
                        marker=marker_map[part],
                        linewidth=2.0,
                        markersize=8,
                        label=f"{group}, {part}",
                        zorder=3,
                    )

        # -----------------------------
        # Optional log scale
        # -----------------------------
        if log_x:
            ax.set_xscale("log")

        # -----------------------------
        # Formatting
        # -----------------------------
        ax.set_title(state_names.get(state, state), fontsize=18, color="black")
        ax.set_xlabel("Total system cost (Million $)", fontsize=16, color="black")
        ax.set_ylabel("RPS target reached (%)", fontsize=16, color="black")

        ax.set_yticks(state_rps_order)
        ax.set_ylim(min(state_rps_order) - 3, max(state_rps_order) + 5)

        ax.tick_params(axis="both", colors="black", labelcolor="black")
        ax.grid(True, which="both", linestyle="--", alpha=0.3, color="gray")

        for spine in ax.spines.values():
            spine.set_color("black")
            spine.set_linewidth(1.0)

    # -----------------------------
    # Legend
    # -----------------------------
    strategy_handles = [
        Line2D([0], [0], color=group_colors["Base only"], marker="o", lw=2, label="Base"),
        Line2D([0], [0], color=group_colors["V1G"], marker="o", lw=2, label="V1G"),
        Line2D([0], [0], color=group_colors["V2G"], marker="o", lw=2, label="V2G"),
        Line2D([0], [0], color=group_colors["V1G+V2G"], marker="o", lw=2, label="V1G+V2G"),
    ]

    participation_handles = [
        Line2D(
            [0], [0],
            color="black",
            marker=marker_map["30%"],
            markersize=8,
            markerfacecolor="gray",
            markeredgecolor="black",
            lw=0,
            label="30% participation",
        ),
        Line2D(
            [0], [0],
            color="black",
            marker=marker_map["50%"],
            markersize=8,
            markerfacecolor="black",
            markeredgecolor="black",
            lw=0,
            label="50% participation",
        ),
    ]

    legend = fig.legend(
        handles=strategy_handles + participation_handles,
        loc="lower center",
        ncol=6,
        frameon=True,
        bbox_to_anchor=(0.5, -0.06),
        fontsize=16,
    )

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    for text in legend.get_texts():
        text.set_color("black")

    scale_label = "log scale" if log_x else "linear scale"

    fig.suptitle(
        f"RPS target reached under different EV charging strategies\n"
        f"Battery capex = ${batt_capex}/kWh, x axis = {scale_label}",
        fontsize=18,
        y=1.01,
        color="black",
    )

    plt.tight_layout()

    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white",
        )

    return df, fig, axes
```

```python
def build_state_results_dirs(state):
    return {
        "flex": [
            f"scenario_results_2030_slow_even_{state}_flex",
            f"scenario_results_2030_mid_even_{state}_flex",
            f"scenario_results_2030_fast_even_{state}_flex",
        ],
        "midnight": [
            f"scenario_results_2030_slow_even_{state}_midnight",
            f"scenario_results_2030_mid_even_{state}_midnight",
            f"scenario_results_2030_fast_even_{state}_midnight",
        ],
        "arrive": [
            f"scenario_results_2030_slow_even_{state}_arrive",
            f"scenario_results_2030_mid_even_{state}_arrive",
            f"scenario_results_2030_fast_even_{state}_arrive",
        ],
        "delay": [
            f"scenario_results_2030_slow_even_{state}_delay",
            f"scenario_results_2030_mid_even_{state}_delay",
            f"scenario_results_2030_fast_even_{state}_delay",
        ],
    }


def load_total_costs_from_results_labeled(results_dir, scenarios=None, scenario_label=None):
    df = load_total_costs_from_results(results_dir, scenarios=scenarios)
    df["scenario_label"] = scenario_label
    return df


def load_all_state_scenario_costs(
    state,
    scenarios=None,
):
    results_dirs_dict = build_state_results_dirs(state)

    df_list = []

    for scenario_label, dirs in results_dirs_dict.items():
        for rd in dirs:
            if not os.path.exists(rd):
                print(f"Skipping missing folder: {rd}")
                continue

            temp = load_total_costs_from_results_labeled(
                results_dir=rd,
                scenarios=scenarios,
                scenario_label=scenario_label,
            )

            temp["state"] = state
            df_list.append(temp)

    if len(df_list) == 0:
        raise ValueError(f"No data found for state: {state}")

    return pd.concat(df_list, ignore_index=True)


def plot_total_cost_delta_vs_rps_all_states(
    states=("CA", "NY", "TX"),
    state_names=None,
    scenarios=None,
    save_path=None,
    batt_order=(150, 250),
    rps_order_by_state=None,
    baseline_scenario_label_by_state=None,
    program_scenario_label_by_state=None,
    selected_participation=("30%", "50%"),
):
    """
    Rows:
        States.

    Columns:
        Battery capex values.

    Main line:
        Mid adoption.

    Shaded area:
        Slow to Fast adoption range.

    Baseline:
        Selected using baseline_scenario_label.

    Program rows:
        V1G, V2G, and V1G+V2G selected using program_scenario_label.
    """

    from matplotlib.lines import Line2D
    from matplotlib.patches import Patch
    if baseline_scenario_label_by_state is None:
        baseline_scenario_label_by_state = {state: "flex" for state in states}

    if program_scenario_label_by_state is None:
        program_scenario_label_by_state = {state: "flex" for state in states}
    if state_names is None:
        state_names = {
            "CA": "California",
            "NY": "New York",
            "TX": "Texas",
        }

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    # -----------------------------
    # Load all states
    # -----------------------------
    df_list = []

    for state in states:
        temp = load_all_state_scenario_costs(
            state=state,
            scenarios=scenarios,
        )
        df_list.append(temp)

    df = pd.concat(df_list, ignore_index=True)

    df["group_plot"] = df["group"].replace({"Base only": "Base"})
    df["batt_capex_num"] = df["batt_capex_label"].str.extract(r"(\d+)").astype(int)
    df["rps_percent"] = (df["rps_ratio"] * 100).round().astype(int)

    # -----------------------------
    # Keep state-specific RPS values
    # -----------------------------
    df = pd.concat(
        [
            df[
                (df["state"] == state) &
                (df["rps_percent"].isin(rps_order_by_state[state]))
            ]
            for state in states
        ],
        ignore_index=True
    )

    # -----------------------------
    # Select baseline and program scenario labels
    # -----------------------------
    base_parts = []
    program_parts = []

    for state in states:
        base_label = baseline_scenario_label_by_state[state]
        program_label = program_scenario_label_by_state[state]

        base_part = df[
            (df["state"] == state) &
            (df["group_plot"] == "Base") &
            (df["scenario_label"] == base_label)
        ].copy()

        program_part = df[
            (df["state"] == state) &
            (df["group_plot"].isin(["V1G", "V2G", "V1G+V2G"])) &
            (df["scenario_label"] == program_label) &
            (df["participation"].isin(selected_participation))
        ].copy()

        base_parts.append(base_part)
        program_parts.append(program_part)

    base_df_raw = pd.concat(base_parts, ignore_index=True)
    program_df_raw = pd.concat(program_parts, ignore_index=True)

    for state in states:
        base_label = baseline_scenario_label_by_state[state]
        program_label = program_scenario_label_by_state[state]

        base_check = base_df_raw[
            (base_df_raw["state"] == state)
        ]

        program_check = program_df_raw[
            (program_df_raw["state"] == state)
        ]

        if base_check.empty:
            raise ValueError(
                f"No baseline rows found for state={state}, baseline_scenario_label={base_label}"
            )

        if program_check.empty:
            raise ValueError(
                f"No program rows found for state={state}, program_scenario_label={program_label}"
            )

    # -----------------------------
    # Build baseline table
    # -----------------------------
    base_df = base_df_raw[[
        "state",
        "adoption",
        "rps_ratio",
        "batt_capex_label",
        "objective_value"
    ]].rename(columns={"objective_value": "base_cost"})

    # -----------------------------
    # Merge baseline into program rows
    # -----------------------------
    df_plot = program_df_raw.merge(
        base_df,
        on=["state", "adoption", "rps_ratio", "batt_capex_label"],
        how="left",
    )

    if df_plot["base_cost"].isna().any():
        missing = df_plot[df_plot["base_cost"].isna()][[
            "state",
            "scenario_id",
            "adoption",
            "rps_ratio",
            "batt_capex_label",
            "group_plot",
            "participation",
        ]]
        raise ValueError(f"Missing matching Base scenario for:\n{missing}")

    df_plot["delta_cost_percent"] = 100 * (
        df_plot["objective_value"] - df_plot["base_cost"]
    ) / df_plot["base_cost"]

    # -----------------------------
    # Plot settings
    # -----------------------------
    group_order = ["V1G", "V2G", "V1G+V2G"]

    color_map = {
        "V1G": "#4E79A7",
        "V2G": "#59A14F",
        "V1G+V2G": "#E15759",
    }

    linestyle_map = {
        "30%": "--",
        "50%": "-",
    }

    marker_map = {
        "30%": "s",
        "50%": "^",
    }

    plt.style.use("default")

    fig, axes = plt.subplots(
        len(states),
        len(batt_order),
        figsize=(14, 4.2 * len(states)),
        sharey=False,
        facecolor="white",
    )

    if len(states) == 1 and len(batt_order) == 1:
        axes = np.array([[axes]])
    elif len(states) == 1:
        axes = np.array([axes])
    elif len(batt_order) == 1:
        axes = np.array([[ax] for ax in axes])

    # -----------------------------
    # Plot
    # -----------------------------
    for row_idx, state in enumerate(states):
        for col_idx, batt in enumerate(batt_order):
            ax = axes[row_idx, col_idx]
            ax.set_facecolor("white")

            subset_state_batt = df_plot[
                (df_plot["state"] == state) &
                (df_plot["batt_capex_num"] == batt)
            ].copy()

            for group in group_order:
                for participation in selected_participation:

                    subset = subset_state_batt[
                        (subset_state_batt["group_plot"] == group) &
                        (subset_state_batt["participation"] == participation)
                    ].copy()

                    if subset.empty:
                        continue

                    pivot = subset.pivot_table(
                        index="rps_percent",
                        columns="adoption",
                        values="delta_cost_percent",
                        aggfunc="mean",
                    ).sort_index()

                    if "Mid" not in pivot.columns:
                        continue

                    x = pivot.index.values
                    y_mid = pivot["Mid"].values

                    # Shaded range from Slow to Fast
                    if ("Slow" in pivot.columns) and ("Fast" in pivot.columns):
                        y_low = pivot[["Slow", "Fast"]].min(axis=1).values
                        y_high = pivot[["Slow", "Fast"]].max(axis=1).values

                        ax.fill_between(
                            x,
                            y_low,
                            y_high,
                            color=color_map[group],
                            alpha=0.14,
                            linewidth=0,
                            zorder=1,
                        )

                    ax.plot(
                        x,
                        y_mid,
                        color=color_map[group],
                        linestyle=linestyle_map[participation],
                        marker=marker_map[participation],
                        linewidth=2.2,
                        markersize=6,
                        zorder=3,
                        label=f"{group} {participation}",
                    )

            ax.axhline(
                0,
                color="black",
                linewidth=1.0,
                linestyle="--",
                zorder=2,
            )

            ax.set_title(
                f"{state_names.get(state, state)}, battery capex = ${batt}/kWh",
                fontsize=18,
                color="black",
            )

            ax.set_xlabel("RPS target (%)", fontsize=17, color="black")
            ax.set_ylabel(
                "Change in total cost\nrelative to Base (%)",
                fontsize=16,
                color="black",
            )

            ax.set_xticks(rps_order_by_state[state])
            ax.grid(True, alpha=0.25, color="gray", linestyle="--", linewidth=0.6)
            ax.tick_params(axis="both", colors="black", labelcolor="black",labelsize=14)

            for spine in ax.spines.values():
                spine.set_color("black")

    # -----------------------------
    # Legend
    # -----------------------------
    strategy_handles = [
        Line2D([0], [0], color=color_map["V1G"], lw=2.2, label="V1G"),
        Line2D([0], [0], color=color_map["V2G"], lw=2.2, label="V2G"),
        Line2D([0], [0], color=color_map["V1G+V2G"], lw=2.2, label="V1G+V2G"),
    ]

    participation_handles = [
        Line2D(
            [0], [0],
            color="black",
            linestyle=linestyle_map["30%"],
            marker=marker_map["30%"],
            lw=2.2,
            label="30% participation",
        ),
        Line2D(
            [0], [0],
            color="black",
            linestyle=linestyle_map["50%"],
            marker=marker_map["50%"],
            lw=2.2,
            label="50% participation",
        ),
    ]

    band_handle = [
        Patch(
            facecolor="gray",
            alpha=0.14,
            edgecolor="none",
            label="Slow to Fast adoption range",
        )
    ]

    legend = fig.legend(
        handles=strategy_handles + participation_handles + band_handle,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.02),
        ncol=6,
        frameon=True,
        fontsize=15,
    )

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    for text in legend.get_texts():
        text.set_color("black")

    baseline_text = ", ".join(
    [f"{state}:{baseline_scenario_label_by_state[state]}" for state in states]
    )

    program_text = ", ".join(
        [f"{state}:{program_scenario_label_by_state[state]}" for state in states]
    )

    fig.suptitle(
        f"Change in total system cost relative to Base scenario\n",
        # f"Baseline → {baseline_text}\n"
        # f"Program → {program_text}",
        fontsize=18,
        y=1,
        color="black",
    )

    plt.tight_layout(rect=[0, 0.04, 1, 0.97])

    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white",
        )

    return df_plot, fig, axes
```

```python
def parse_scenario_metadata(folder_name):
    meta = {}

    m = re.search(r"^s(\d+)_", folder_name)
    meta["scenario_id"] = int(m.group(1)) if m else None

    # NEW: Extract v1g_share and v2g_share from folder name
    m_v1g = re.search(r"v1g(\d+)", folder_name, re.IGNORECASE)
    m_v2g = re.search(r"v2g(\d+)", folder_name, re.IGNORECASE)

    v1g_val = int(m_v1g.group(1)) if m_v1g else 0
    v2g_val = int(m_v2g.group(1)) if m_v2g else 0

    meta["v1g_share"] = v1g_val / 100.0
    meta["v2g_share"] = v2g_val / 100.0

    # Determine participation level
    total_participation = v1g_val + v2g_val
    if total_participation == 0:
        meta["participation"] = "0%"
        meta["group"] = "Base only"
    elif total_participation == 10:
        meta["participation"] = "10%"
    elif total_participation == 30:
        meta["participation"] = "30%"
    else:
        meta["participation"] = f"{total_participation}%"

    # Determine group
    if "_Base_only_" in folder_name:
        meta["group"] = "Base only"
    elif v1g_val > 0 and v2g_val > 0:
        meta["group"] = "V1G+V2G"
    elif v1g_val > 0:
        meta["group"] = "V1G"
    elif v2g_val > 0:
        meta["group"] = "V2G"
    else:
        meta["group"] = "Base only"

    m = re.search(r"rps(\d+)", folder_name)
    meta["rps"] = int(m.group(1)) if m else None

    m = re.search(r"bcapex(\d+)", folder_name)
    meta["batt_capex"] = int(m.group(1)) if m else None

    return meta
def get_label_for_state(label_input, state):
    """
    Allows either:
        "flex"
    or:
        {"CA": "flex", "NY": "arrive", "TX": "midnight"}
    """
    if isinstance(label_input, dict):
        return label_input[state]
    return label_input


def get_graph_path_for_state(graph_json_path_by_state, state):
    """
    Allows either:
        one graph path string
    or:
        {"CA": "...", "NY": "...", "TX": "..."}
    """
    if isinstance(graph_json_path_by_state, dict):
        return graph_json_path_by_state[state]
    return graph_json_path_by_state

def map_fuel_for_plot_extended(fuel):
    f = str(fuel).strip().lower()

    if f == "battery_discharge":
        return "Battery Discharge"
    elif f == "v2g_discharge":
        return "V2G Discharge"
    elif f == "battery_charge":
        return "Battery Charge"
    elif f == "v2g_charge":
        return "V2G Charge"
    elif f == "solar":
        return "Solar"
    elif f == "wind":
        return "Wind"
    elif f == "nuclear":
        return "Nuclear"
    elif f == "coal":
        return "Coal"
    elif f == "natural gas combined cycle":
        return "CCNG"
    elif f == "natural gas turbine":
        return "Gas\nTurbine"
    elif f == "hydro":
        return "Hydro"
    elif f == "geothermal":
        return "Geothermal"
    elif f == "biomass":
        return "Biomass"
    elif f == "waste":
        return "Waste"
    elif f == "import":
        return "Import"
    elif f == "oil":
        return "Steam"
    elif f == "non-fossil":
        return "Non-fossil"
    else:
        return "Other"

def generation_summary_from_solution_json(solution_json_path, graph, time_step_h=1.0):
    import json
    import numpy as np
    import pandas as pd

    with open(solution_json_path) as f:
        data = json.load(f)

    rows = []

    for node in data["nodes"]:
        region = node["id"]

        if region not in graph._node:
            continue

        for handle, a_sol in node["assets"].items():

            if handle not in graph._node[region]["assets"]:
                continue

            meta = graph._node[region]["assets"][handle]

            cls = meta.get("_class", "")
            typ = meta.get("type", "")
            fuel = meta.get("fuel", None) or typ or cls

            # -------------------------
            # 1. Skip real demand
            # -------------------------
            if cls == "Load" and typ == "load":
                continue

            # -------------------------
            # 2. Storage → use production only
            # -------------------------
            if cls == "Store":
                prod = np.array(a_sol.get("production", []), dtype=float)

                if prod.size == 0:
                    continue

                handle_lower = str(handle).lower()
                type_lower = str(meta.get("type", "")).lower()
                fuel_lower = str(meta.get("fuel", "")).lower()

                is_v2g = (
                    type_lower == "ev_v2g"
                    or fuel_lower == "ev_v2g"
                    or "ev_v2g" in handle_lower
                )

                fuel_name = "v2g_discharge" if is_v2g else "battery_discharge"

                energy_gwh = prod.sum() * time_step_h / 1e9

                rows.append({
                    "resource": map_fuel_for_plot_extended(fuel_name),
                    "energy_gwh": energy_gwh
                })

                continue  # IMPORTANT: avoid double counting

            # -------------------------
            # 3. All other generators (including solar/wind)
            # -------------------------
            net = np.array(a_sol.get("net", []), dtype=float)

            if net.size == 0:
                continue

            gen = np.clip(net, 0, None)

            energy_gwh = gen.sum() * time_step_h / 1e9

            rows.append({
                "resource": map_fuel_for_plot_extended(fuel),
                "energy_gwh": energy_gwh
            })

    if len(rows) == 0:
        raise ValueError(f"No generation data found in {solution_json_path}")

    df = pd.DataFrame(rows)
    df = df.groupby("resource", as_index=False)["energy_gwh"].sum()

    return df


def collect_generation_delta_for_state_group(
    state,
    adoption,
    graph_json_path,
    batt_capex_target=150,
    baseline_scenario_label="flex",
    program_scenario_label="flex",
    selected_group="V2G",
    selected_participation=("30%", "50%"),
    time_step_h=1.0,
):
    """
    Collect generation delta for one state and one selected program group.

    selected_group:
        "V1G"
        "V2G"
        "V1G+V2G"
    """

    graph = good.graph.graph_from_json(graph_json_path)

    baseline_dir = f"scenario_results_2030_{adoption}_even_{state}_{baseline_scenario_label}"
    program_dir = f"scenario_results_2030_{adoption}_even_{state}_{program_scenario_label}"

    all_rows = []

    # -----------------------------
    # Baseline rows
    # -----------------------------
    for folder in sorted(glob.glob(os.path.join(baseline_dir, "s*"))):
        folder_name = os.path.basename(folder)
        meta = parse_scenario_metadata(folder_name)

        if meta["batt_capex"] != batt_capex_target:
            continue

        if meta["group"] != "Base only":
            continue

        solution_jsons = glob.glob(os.path.join(folder, "*_solution.json"))
        if not solution_jsons:
            continue

        gen = generation_summary_from_solution_json(
            solution_json_path=solution_jsons[0],
            graph=graph,
            time_step_h=1.0,
        )

        gen["state"] = state
        gen["scenario_id"] = meta["scenario_id"]
        gen["group"] = meta["group"]
        gen["participation"] = meta["participation"]
        gen["rps"] = meta["rps"]
        gen["batt_capex"] = meta["batt_capex"]
        gen["scenario_label"] = baseline_scenario_label

        all_rows.append(gen)

    # -----------------------------
    # Program rows
    # -----------------------------
    for folder in sorted(glob.glob(os.path.join(program_dir, "s*"))):
        folder_name = os.path.basename(folder)
        meta = parse_scenario_metadata(folder_name)

        if meta["batt_capex"] != batt_capex_target:
            continue

        if meta["group"] != selected_group:
            continue

        if meta["participation"] not in selected_participation:
            continue

        solution_jsons = glob.glob(os.path.join(folder, "*_solution.json"))
        if not solution_jsons:
            continue

        gen = generation_summary_from_solution_json(
            solution_json_path=solution_jsons[0],
            graph=graph,
            time_step_h=1.0,
        )

        gen["state"] = state
        gen["scenario_id"] = meta["scenario_id"]
        gen["group"] = meta["group"]
        gen["participation"] = meta["participation"]
        gen["rps"] = meta["rps"]
        gen["batt_capex"] = meta["batt_capex"]
        gen["scenario_label"] = program_scenario_label

        all_rows.append(gen)

    if not all_rows:
        raise ValueError(
            f"No valid generation data found for state={state}, group={selected_group}"
        )

    df = pd.concat(all_rows, ignore_index=True)

    base = df[df["group"] == "Base only"].copy()
    base = base.rename(columns={"energy_gwh": "base_energy_gwh"})
    base = base[["state", "rps", "batt_capex", "resource", "base_energy_gwh"]]

    comp = df[df["group"] == selected_group].copy()

    merged = comp.merge(
        base,
        on=["state", "rps", "batt_capex", "resource"],
        how="left",
    )

    merged["base_energy_gwh"] = merged["base_energy_gwh"].fillna(0.0)
    merged["delta_gwh"] = merged["energy_gwh"] - merged["base_energy_gwh"]

    return merged


def plot_generation_delta_all_states_by_rps_1x3(
    states=("CA", "NY", "TX"),
    state_names=None,
    graph_json_path_by_state=None,
    adoption="mid",
    batt_capex_target=150,
    baseline_scenario_label_by_state="flex",
    program_scenario_label_by_state="flex",
    selected_group="V2G",
    selected_participation=("30%", "50%"),
    rps_order_by_state=None,
    save_path=None,
    label_threshold_by_state=None,
):
    import os
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    from matplotlib.patches import Patch

    if state_names is None:
        state_names = {
            "CA": "California",
            "NY": "New York",
            "TX": "Texas",
        }
    if label_threshold_by_state is None:
        label_threshold_by_state = {
            "CA": 20,
            "NY": 20,
            "TX": 20,
        }

    if graph_json_path_by_state is None:
        graph_json_path_by_state = {
            "CA": "Examples/California_new.json",
            "NY": "Examples/NewYork_new.json",
            "TX": "Examples/Texas_new.json",
        }

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    resource_order = [
        "Solar", "Wind", "Nuclear", "Coal", "CCNG", "Gas\nTurbine",
        "Hydro", "Geothermal", "Biomass", "Waste", "Import",
        "Steam",   # ← replace Oil
        "Non-fossil", "Other", "Battery Discharge", "V2G Discharge",
    ]

    colors = {
        "Solar": "#f4b400",
        "Wind": "#4a90e2",
        "Nuclear": "#ff1f1f",
        "Coal": "#000000",
        "CCNG": "#8c8c8c",
        "Gas\nTurbine": "#5f5f5f",
        "Hydro": "#4fc3f7",
        "Geothermal": "#8e44ad",
        "Biomass": "#2e7d32",
        "Waste": "#66bb6a",
        "Import": "#795548",
        "Steam": "#b71c1c",
        "Non-fossil": "#00acc1",
        "Other": "#9e9e9e",
        "Battery Discharge": "#6d4c41",
        "V2G Discharge": "#c2185b",
    }

    # -----------------------------
    # Load data
    # -----------------------------
    df_list = []

    for state in states:
        graph_path = get_graph_path_for_state(graph_json_path_by_state, state)
        base_label = get_label_for_state(baseline_scenario_label_by_state, state)
        program_label = get_label_for_state(program_scenario_label_by_state, state)

        temp = collect_generation_delta_for_state_group(
            state=state,
            adoption=adoption,
            graph_json_path=graph_path,
            batt_capex_target=batt_capex_target,
            baseline_scenario_label=base_label,
            program_scenario_label=program_label,
            selected_group=selected_group,
            selected_participation=selected_participation,
        )

        df_list.append(temp)

    df = pd.concat(df_list, ignore_index=True)

    df = pd.concat(
        [
            df[
                (df["state"] == state) &
                (df["rps"].isin(rps_order_by_state[state]))
            ]
            for state in states
        ],
        ignore_index=True,
    )

    present_resources = [
        r for r in resource_order
        if r in df["resource"].unique()
    ]

    # -----------------------------
    # Figure
    # -----------------------------
    fig, axes = plt.subplots(
        1,
        len(states),
        figsize=(18, 10),
        sharey=False,
        facecolor="white",
    )

    if len(states) == 1:
        axes = [axes]

    bar_width = 1.2
    participation_gap = 0.30
    rps_gap = 0.85

    resources_in_legend = set()

    for ax, state in zip(axes, states):
        state_label_threshold = label_threshold_by_state.get(state, 20)
        ax.set_facecolor("white")

        state_rps_values = rps_order_by_state[state]
        sub_state = df[df["state"] == state].copy()

        x_map = {}
        xticks = []
        xticklabels = []
        rps_centers = []
        separator_positions = []

        current_x = 0.0

        for rps_idx, rps in enumerate(state_rps_values):
            local_positions = []

            for p in selected_participation:
                x_map[(rps, p)] = current_x
                xticks.append(current_x)
                xticklabels.append(p)
                local_positions.append(current_x)

                current_x += bar_width + participation_gap

            rps_centers.append(np.mean(local_positions))

            if rps_idx < len(state_rps_values) - 1:
                sep = current_x + rps_gap / 2 - participation_gap
                separator_positions.append(sep)

            current_x += rps_gap

        ordered_keys = [
            (rps, p)
            for rps in state_rps_values
            for p in selected_participation
        ]

        pos_bottom = {key: 0.0 for key in ordered_keys}
        neg_bottom = {key: 0.0 for key in ordered_keys}

        for resource in present_resources:
            for key in ordered_keys:
                rps, participation = key
                x = x_map[key]

                temp = sub_state[
                    (sub_state["rps"] == rps) &
                    (sub_state["participation"] == participation) &
                    (sub_state["resource"] == resource)
                ]

                val = temp["delta_gwh"].sum() if not temp.empty else 0.0

                if abs(val) > 1e-10:
                    resources_in_legend.add(resource)

                if val >= 0:
                    bottom = pos_bottom[key]

                    ax.bar(
                        x,
                        val,
                        width=bar_width,
                        bottom=bottom,
                        color=colors.get(resource, "#9e9e9e"),
                        edgecolor="none",
                    )

                    if abs(val) >= state_label_threshold:
                        ax.text(
                            x,
                            bottom + val / 2,
                            resource,
                            ha="center",
                            va="center",
                            fontsize=12,
                            color="black",
                            # fontweight="bold",
                        )

                    pos_bottom[key] += val

                else:
                    bottom = neg_bottom[key]

                    ax.bar(
                        x,
                        val,
                        width=bar_width,
                        bottom=bottom,
                        color=colors.get(resource, "#9e9e9e"),
                        edgecolor="none",
                    )

                    if abs(val) >= state_label_threshold:
                        ax.text(
                            x,
                            bottom + val / 2,
                            resource,
                            ha="center",
                            va="center",
                            fontsize=12,
                            color="black",
                            # fontweight="bold",
                        )

                    neg_bottom[key] += val

        # zero line
        ax.axhline(0, color="black", linewidth=1.1)

        # RPS separators
        for xpos in separator_positions:
            ax.axvline(
                xpos,
                color="black",
                linestyle="--",
                linewidth=1.0,
                alpha=0.75,
            )

        # participation labels
        ax.set_xticks(xticks)
        ax.set_xticklabels(
            xticklabels,
            fontsize=16,
            fontweight="bold",
            color="black",
        )

        # Get y-axis range before placing custom labels
        ymin, ymax = ax.get_ylim()
        y_range = ymax - ymin

        # selected scenario label at bottom
        scenario_label_y = ymin - 0.10 * y_range

        ax.text(
            np.mean(xticks),
            scenario_label_y,
            selected_group,
            ha="center",
            va="top",
            fontsize=16,
            fontweight="bold",
            color="black",
            clip_on=False,
        )

        # RPS labels above plot
        rps_label_y = ymax + 0.01 * y_range

        for rps, center in zip(state_rps_values, rps_centers):
            ax.text(
                center,
                rps_label_y,
                f"RPS {rps}%",
                ha="center",
                va="bottom",
                fontsize=16,
                # fontweight="bold",
                color="black",
                clip_on=False,
            )


        ax.set_title(
            state_names.get(state, state),
            fontsize=18,
            fontweight="bold",
            color="black",
            pad=25,   # ← increase vertical distance
        )

        ax.set_ylabel(
            "Change in generation (GWh)",
            fontsize=18,
            color="black",
        )

        ax.grid(axis="y", color="gray", alpha=0.25, linestyle="--")
        ax.tick_params(axis="both", colors="black", labelsize=14, labelcolor="black")

        for spine in ax.spines.values():
            spine.set_color("black")

        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    # -----------------------------
    # Legend
    # -----------------------------
    legend_resources = [
        r for r in present_resources
        if r in resources_in_legend
    ]

    legend_elements = [
        Patch(
            facecolor=colors.get(resource, "#9e9e9e"),
            edgecolor="black",
            label=resource,
        )
        for resource in legend_resources
    ]

    legend = fig.legend(
        handles=legend_elements,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.01),
        ncol=min(len(legend_elements), 8),
        frameon=True,
        fontsize=16,
        title="Generation resources",
    )
    # Change title size
    legend.get_title().set_fontsize(18)   # ← adjust here
    legend.get_title().set_weight("bold") # optional
    legend.get_title().set_color("black") # optional

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    for text in legend.get_texts():
        text.set_color("black")

    if legend.get_title() is not None:
        legend.get_title().set_color("black")

    baseline_text = ", ".join(
        [
            f"{state}:{get_label_for_state(baseline_scenario_label_by_state, state)}"
            for state in states
        ]
    )

    program_text = ", ".join(
        [
            f"{state}:{get_label_for_state(program_scenario_label_by_state, state)}"
            for state in states
        ]
    )

    fig.suptitle(
        f"Generation mix change for {selected_group} relative to Base\n"
        f"Battery capex = ${batt_capex_target}/kWh, adoption = {adoption}\n",
        # f"Base = {baseline_text}; Program = {program_text}",
        fontsize=19,
        y=0.90,
        color="black",
    )

    plt.tight_layout(rect=[0, 0.10, 1, 0.90])

    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white",
        )

    return df, fig, axes


df_gen_1x3, fig, axes = plot_generation_delta_all_states_by_rps_1x3(
    states=("CA", "NY", "TX"),
    adoption="mid",
    batt_capex_target=250,
    selected_group="V2G",
    selected_participation=("30%", "50%"),
    baseline_scenario_label_by_state={
        "CA": "midnight",
        "NY": "midnight",
        "TX": "arrive",
    },
    program_scenario_label_by_state={
        "CA": "flex",
        "NY": "flex",
        "TX": "midnight",
    },
    graph_json_path_by_state={
        "CA": "Examples/California_new.json",
        "NY": "Examples/New_York.json",
        "TX": "Examples/Texas.json",
    },
    rps_order_by_state={
        "CA": [50, 60, 70],
        "NY": [60, 70, 80],
        "TX": [50, 60, 70],
    },
    label_threshold_by_state={
        "CA": 20,
        "NY": 5,
        "TX": 50,
    },
    save_path="figures/generation_delta_1x3_V2G_mid_bcapex150.png",
)
```

```python
import os
import re
import glob
import json
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# 1. Read scenario JSON
# ============================================================

SCENARIOS_BY_STATE = {
    "CA": json.load(open("Examples/scenarios_2030_CA.json")),
    "NY": json.load(open("Examples/scenarios_2030_NY.json")),
    "TX": json.load(open("Examples/scenarios_2030_TX.json")),
}


# ============================================================
# 2. Helpers
# ============================================================

def get_label_for_state(label_input, state):
    if isinstance(label_input, dict):
        return label_input[state]
    return label_input


def get_scenario_id_from_folder(folder_name):
    """
    Example:
    s01_s01_fast_m7_d7 -> 1
    s41_s41_fast_m7_d7 -> 41
    """
    m = re.search(r"^s(\d+)_", folder_name)
    if m is None:
        return None
    return int(m.group(1))


def get_scenario_cfg(SCENARIOS, scenario_id):
    """
    Handles both JSON key styles:
    "1" or "s01" or integer-like keys.
    """
    possible_keys = [
        str(scenario_id),
        f"s{scenario_id:02d}",
        f"s{scenario_id}",
        scenario_id,
    ]

    for key in possible_keys:
        if key in SCENARIOS:
            return SCENARIOS[key]

    raise KeyError(f"Scenario {scenario_id} not found in SCENARIOS")


def batt_capex_to_kwh(value):
    """
    If batt_capex_cost is stored as $/J, convert to $/kWh.
    If it is already like 150 or 250, keep it.
    """
    if value < 1:
        return int(round(value * 3.6e9 / 1000))
    return int(round(value))


def metadata_from_json(SCENARIOS, scenario_id):
    cfg = get_scenario_cfg(SCENARIOS, scenario_id)

    rps = int(round(cfg["rps_ratio"] * 100))
    batt_capex = batt_capex_to_kwh(cfg["batt_capex_cost"])

    v1g_share = cfg.get("v1g_share", 0)
    v2g_share = cfg.get("v2g_share", 0)

    v1g_val = int(round(v1g_share * 100))
    v2g_val = int(round(v2g_share * 100))

    total_participation = v1g_val + v2g_val

    if total_participation == 0:
        group = "Base only"
    elif v1g_val > 0 and v2g_val > 0:
        group = "V1G+V2G"
    elif v1g_val > 0:
        group = "V1G"
    elif v2g_val > 0:
        group = "V2G"
    else:
        group = "Base only"

    return {
        "scenario_id": scenario_id,
        "rps": rps,
        "batt_capex": batt_capex,
        "v1g_share": v1g_share,
        "v2g_share": v2g_share,
        "participation": f"{total_participation}%",
        "group": group,
    }


def map_capex_resource(fuel):
    f = str(fuel).strip().lower()

    if f == "solar":
        return "Solar"
    if f == "wind":
        return "Wind"
    if f == "battery":
        return "Battery"

    return "Other"


def capex_summary_from_capex_summary_csv(capex_summary_path):
    df = pd.read_csv(capex_summary_path)

    df = df.copy()
    df["fuel_clean"] = df["fuel"].astype(str).str.lower().str.strip()

    resource_map = {
        "solar": "Solar",
        "wind": "Wind",
        "battery": "Battery",
    }

    df = df[df["fuel_clean"].isin(resource_map.keys())].copy()
    df["resource"] = df["fuel_clean"].map(resource_map)

    out = (
        df.groupby("resource", as_index=False)["capex_added_MW"]
        .sum()
    )

    return out


# ============================================================
# 3. Collect capex delta
# ============================================================

def collect_capex_delta_for_state_group(
    state,
    adoption,
    SCENARIOS,
    batt_capex_target=150,
    baseline_scenario_label="arrive",
    program_scenario_label="flex",
    selected_group="V2G",
    selected_participation=("30%", "50%"),
):
    baseline_dir = f"scenario_results_2030_{adoption}_even_{state}_{baseline_scenario_label}"
    program_dir = f"scenario_results_2030_{adoption}_even_{state}_{program_scenario_label}"

    all_rows = []

    # -----------------------------
    # Baseline
    # -----------------------------
    for folder in sorted(glob.glob(os.path.join(baseline_dir, "s*"))):
        folder_name = os.path.basename(folder)
        scenario_id = get_scenario_id_from_folder(folder_name)

        if scenario_id is None:
            continue

        meta = metadata_from_json(SCENARIOS, scenario_id)

        if meta["batt_capex"] != batt_capex_target:
            continue

        if meta["group"] != "Base only":
            continue

        capex_files = glob.glob(os.path.join(folder, "*_capex_summary.csv"))

        if not capex_files:
            continue

        cap = capex_summary_from_capex_summary_csv(capex_files[0])

        for key, value in meta.items():
            cap[key] = value

        cap["state"] = state
        cap["scenario_label"] = baseline_scenario_label

        all_rows.append(cap)

    # -----------------------------
    # Program
    # -----------------------------
    for folder in sorted(glob.glob(os.path.join(program_dir, "s*"))):
        folder_name = os.path.basename(folder)
        scenario_id = get_scenario_id_from_folder(folder_name)

        if scenario_id is None:
            continue

        meta = metadata_from_json(SCENARIOS, scenario_id)

        if meta["batt_capex"] != batt_capex_target:
            continue

        if meta["group"] != selected_group:
            continue

        if meta["participation"] not in selected_participation:
            continue

        capex_files = glob.glob(os.path.join(folder, "*_capex_summary.csv"))

        if not capex_files:
            continue

        cap = capex_summary_from_capex_summary_csv(capex_files[0])

        for key, value in meta.items():
            cap[key] = value

        cap["state"] = state
        cap["scenario_label"] = program_scenario_label

        all_rows.append(cap)

    if not all_rows:
        raise ValueError(
            f"No valid capex summary data found. Check paths and filters:\n"
            f"baseline_dir = {baseline_dir}\n"
            f"program_dir = {program_dir}\n"
            f"state = {state}\n"
            f"adoption = {adoption}\n"
            f"group = {selected_group}\n"
            f"battery capex = {batt_capex_target}\n"
            f"participation = {selected_participation}"
        )

    df = pd.concat(all_rows, ignore_index=True)

    base = df[df["group"] == "Base only"].copy()
    base = base.rename(columns={"capex_added_MW": "base_capex_mw"})

    base = base[
        ["state", "rps", "batt_capex", "resource", "base_capex_mw"]
    ]

    comp = df[df["group"] == selected_group].copy()

    merged = comp.merge(
        base,
        on=["state", "rps", "batt_capex", "resource"],
        how="left",
    )

    merged["base_capex_mw"] = merged["base_capex_mw"].fillna(0.0)
    merged["delta_mw"] = merged["capex_added_MW"] - merged["base_capex_mw"]

    return merged


# ============================================================
# 4. Plot
# ============================================================

def plot_capex_delta_all_states_by_rps_1x3(
    states=("CA",),
    state_names=None,
    adoption="fast",
    SCENARIOS=None,
    batt_capex_target=150,
    baseline_scenario_label_by_state="arrive",
    program_scenario_label_by_state="flex",
    selected_group="V2G",
    selected_participation=("30%", "50%"),
    rps_order_by_state=None,
    save_path=None,
    label_threshold_by_state=None,
):
    import numpy as np

    if SCENARIOS is None:
        raise ValueError("You must pass SCENARIOS.")

    if state_names is None:
        state_names = {
            "CA": "California",
            "NY": "New York",
            "TX": "Texas",
        }

    if label_threshold_by_state is None:
        label_threshold_by_state = {
            "CA": 200,
            "NY": 50,
            "TX": 200,
        }

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    resource_order = ["Solar", "Wind", "Battery"]

    colors = {
        "Solar": "#f4b400",
        "Wind": "#4a90e2",
        "Battery": "#ff1f1f", # purple (better contrast)
    }

    df_list = []

    for state in states:
        base_label = get_label_for_state(
            baseline_scenario_label_by_state,
            state,
        )

        program_label = get_label_for_state(
            program_scenario_label_by_state,
            state,
        )

        temp = collect_capex_delta_for_state_group(
            state=state,
            adoption=adoption,
            SCENARIOS=SCENARIOS[state],
            batt_capex_target=batt_capex_target,
            baseline_scenario_label=base_label,
            program_scenario_label=program_label,
            selected_group=selected_group,
            selected_participation=selected_participation,
        )

        df_list.append(temp)

    df = pd.concat(df_list, ignore_index=True)

    df = pd.concat(
        [
            df[
                (df["state"] == state)
                & (df["rps"].isin(rps_order_by_state[state]))
            ]
            for state in states
        ],
        ignore_index=True,
    )

    present_resources = [
        r for r in resource_order
        if r in df["resource"].unique()
    ]

    fig, axes = plt.subplots(
        1,
        len(states),
        figsize=(7 * len(states), 8),
        sharey=False,
        facecolor="white",
    )

    if len(states) == 1:
        axes = [axes]

    bar_width = 1.2
    participation_gap = 0.30
    rps_gap = 0.85

    resources_in_legend = set()

    for ax, state in zip(axes, states):
        state_label_threshold = label_threshold_by_state.get(state, 200)

        ax.set_facecolor("white")

        state_rps_values = rps_order_by_state[state]
        sub_state = df[df["state"] == state].copy()

        x_map = {}
        xticks = []
        xticklabels = []
        rps_centers = []
        separator_positions = []

        current_x = 0.0

        for rps_idx, rps in enumerate(state_rps_values):
            local_positions = []

            for p in selected_participation:
                x_map[(rps, p)] = current_x
                xticks.append(current_x)
                xticklabels.append(p)
                local_positions.append(current_x)

                current_x += bar_width + participation_gap

            rps_centers.append(np.mean(local_positions))

            if rps_idx < len(state_rps_values) - 1:
                sep = current_x + rps_gap / 2 - participation_gap
                separator_positions.append(sep)

            current_x += rps_gap

        ordered_keys = [
            (rps, p)
            for rps in state_rps_values
            for p in selected_participation
        ]

        pos_bottom = {key: 0.0 for key in ordered_keys}
        neg_bottom = {key: 0.0 for key in ordered_keys}

        for resource in present_resources:
            for key in ordered_keys:
                rps, participation = key
                x = x_map[key]

                temp = sub_state[
                    (sub_state["rps"] == rps)
                    & (sub_state["participation"] == participation)
                    & (sub_state["resource"] == resource)
                ]

                val = temp["delta_mw"].sum() if not temp.empty else 0.0

                if abs(val) > 1e-10:
                    resources_in_legend.add(resource)

                if val >= 0:
                    bottom = pos_bottom[key]
                else:
                    bottom = neg_bottom[key]

                ax.bar(
                    x,
                    val,
                    width=bar_width,
                    bottom=bottom,
                    color=colors.get(resource, "#9e9e9e"),
                    edgecolor="none",
                )

                if abs(val) >= state_label_threshold:
                    ax.text(
                        x,
                        bottom + val / 2,
                        resource,
                        ha="center",
                        va="center",
                        fontsize=12,
                        color="black",
                    )

                if val >= 0:
                    pos_bottom[key] += val
                else:
                    neg_bottom[key] += val

        ax.axhline(0, color="black", linewidth=1.1)

        for xpos in separator_positions:
            ax.axvline(
                xpos,
                color="black",
                linestyle="--",
                linewidth=1.0,
                alpha=0.75,
            )

        ax.set_xticks(xticks)

        ax.set_xticklabels(
            xticklabels,
            fontsize=15,
            fontweight="bold",
            color="black",
        )

        ymin, ymax = ax.get_ylim()
        y_range = ymax - ymin if ymax != ymin else 1

        scenario_label_y = ymin - 0.10 * y_range

        ax.text(
            np.mean(xticks),
            scenario_label_y,
            selected_group,
            ha="center",
            va="top",
            fontsize=16,
            fontweight="bold",
            color="black",
            clip_on=False,
        )

        rps_label_y = ymax + 0.01 * y_range

        for rps, center in zip(state_rps_values, rps_centers):
            ax.text(
                center,
                rps_label_y,
                f"RPS {rps}%",
                ha="center",
                va="bottom",
                fontsize=15,
                color="black",
                clip_on=False,
            )

        ax.set_title(
            state_names.get(state, state),
            fontsize=18,
            fontweight="bold",
            color="black",
            pad=25,
        )

        ax.set_ylabel(
            "Change in added capacity (MW)",
            fontsize=17,
            color="black",
        )

        ax.grid(axis="y", color="gray", alpha=0.25, linestyle="--")
        ax.tick_params(axis="both", colors="black", labelsize=13)

        for spine in ax.spines.values():
            spine.set_color("black")

        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    legend_resources = [
        r for r in present_resources
        if r in resources_in_legend
    ]

    legend_elements = [
    Patch(
        facecolor=colors[r],
        edgecolor="black",
        label=r,
    )
    for r in ["Solar", "Wind", "Battery"]
    ]

    legend = fig.legend(
        handles=legend_elements,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.02),
        ncol=3,
        frameon=True,
        fontsize=14,
        title="Capacity expansion resources",
    )

    legend.get_title().set_fontsize(15)
    legend.get_title().set_weight("bold") # optional
    legend.get_title().set_color("black") # optional

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    for text in legend.get_texts():
        text.set_color("black")

    if legend.get_title() is not None:
        legend.get_title().set_color("black")

    fig.suptitle(
        f"Capacity expansion change for {selected_group} relative to Base\n"
        f"Battery capex = ${batt_capex_target}/kWh, adoption = {adoption}",
        fontsize=18,
        y=0.93,
        color="black",
    )

    plt.tight_layout(rect=[0, 0.10, 1, 0.90])

    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)

        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white",
        )

    return df, fig, axes

df_capex_1x3, fig, axes = plot_capex_delta_all_states_by_rps_1x3(
    states=("CA","NY","TX"),
    adoption="mid",
    SCENARIOS=SCENARIOS_BY_STATE,
    batt_capex_target=250,
    selected_group="V2G",
    selected_participation=("30%", "50%"),
    baseline_scenario_label_by_state={
        "CA": "midnight",
        "NY": "midnight",
        "TX": "arrive",
    },
    program_scenario_label_by_state={
        "CA": "flex",
        "NY": "flex",
        "TX": "midnight",
    },
    rps_order_by_state={
        "CA": [50, 60, 70],
        "NY": [60, 70, 80],
        "TX": [50, 60, 70],
    },
    label_threshold_by_state={
        "CA": 200,
        "NY": 100,
        "TX": 500,
    },
    save_path="figures/capex_delta_CA_V2G_fast_bcapex150.png",
)
```

```python
import os
import re
import glob
import json
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D


# ============================================================
# Load scenarios
# ============================================================

with open("Examples/scenarios_2030_CA.json", "r") as f:
    SCENARIOS = json.load(f)


# ============================================================
# Helpers
# ============================================================

def get_label_for_state(label_input, state):
    if isinstance(label_input, dict):
        return label_input[state]
    return label_input


def get_scenario_id_from_folder_or_file(fp):
    folder = os.path.basename(os.path.dirname(fp))
    fname = os.path.basename(fp)

    m = re.search(r"^s(\d+)_", folder)
    if m:
        return int(m.group(1))

    m = re.search(r"^s(\d+)_", fname)
    if m:
        return int(m.group(1))

    return None


def get_scenario_cfg(SCENARIOS, scenario_id):
    for key in [str(scenario_id), f"s{scenario_id:02d}", f"s{scenario_id}", scenario_id]:
        if key in SCENARIOS:
            return SCENARIOS[key]

    raise KeyError(f"Scenario {scenario_id} not found in SCENARIOS")


def batt_capex_to_kwh(value):
    if value < 1:
        return int(round(value * 3.6e9 / 1000))
    return int(round(value))


def get_rps_from_folder_or_file(fp):
    folder = os.path.basename(os.path.dirname(fp))
    fname = os.path.basename(fp)
    text = folder + "_" + fname

    m = re.search(r"rps(\d+)", text, re.IGNORECASE)
    if m:
        return int(m.group(1))

    return None


def metadata_from_json(SCENARIOS, scenario_id, fp=None):
    cfg = get_scenario_cfg(SCENARIOS, scenario_id)

    rps_from_name = get_rps_from_folder_or_file(fp) if fp is not None else None

    if rps_from_name is not None:
        rps_percent = rps_from_name
    else:
        rps_percent = int(round(cfg["rps_ratio"] * 100))

    batt_capex = batt_capex_to_kwh(cfg["batt_capex_cost"])

    v1g_share = cfg.get("v1g_share", 0)
    v2g_share = cfg.get("v2g_share", 0)

    v1g_val = int(round(v1g_share * 100))
    v2g_val = int(round(v2g_share * 100))
    total_participation = v1g_val + v2g_val

    if total_participation == 0:
        group = "Base only"
    elif v1g_val > 0 and v2g_val > 0:
        group = "V1G+V2G"
    elif v1g_val > 0:
        group = "V1G"
    elif v2g_val > 0:
        group = "V2G"
    else:
        group = "Base only"

    return {
        "scenario_id": scenario_id,
        "rps_percent": rps_percent,
        "rps_ratio": rps_percent / 100,
        "batt_capex_num": batt_capex,
        "v1g_share": v1g_share,
        "v2g_share": v2g_share,
        "participation": f"{total_participation}%",
        "group": group,
    }


def build_one_results_dir(state, adoption_filter, scenario_label):
    adoption = adoption_filter.lower()
    return f"scenario_results_2030_{adoption}_even_{state}_{scenario_label}"


# ============================================================
# Wastage reader from solution JSON
# ============================================================

def wastage_gwh_from_solution_json(solution_json_path, time_step_h=1.0):
    with open(solution_json_path, "r") as f:
        data = json.load(f)

    total_wastage_gwh = 0.0

    for node in data.get("nodes", []):
        wastage = node.get("wastage", None)

        if wastage is None:
            continue

        if isinstance(wastage, list):
            total_wastage_gwh += sum(wastage) * time_step_h / 1e9
        else:
            total_wastage_gwh += float(wastage) * time_step_h / 1e9

    return total_wastage_gwh


def load_wastage_from_results_json(
    results_dir,
    SCENARIOS,
    scenario_label,
    adoption_filter,
    time_step_h=1.0,
):
    files = sorted(
        glob.glob(os.path.join(results_dir, "s*", "*_solution.json"))
    )

    print(f"Found {len(files)} solution JSON files in {results_dir}")

    records = []

    for fp in files:
        try:
            scenario_id = get_scenario_id_from_folder_or_file(fp)

            if scenario_id is None:
                continue

            meta = metadata_from_json(SCENARIOS, scenario_id, fp=fp)

            wastage_gwh = wastage_gwh_from_solution_json(
                fp,
                time_step_h=time_step_h,
            )

            row = meta.copy()
            row["wastage_gwh"] = wastage_gwh
            row["adoption"] = adoption_filter.capitalize()
            row["scenario_label"] = scenario_label
            row["file_path"] = fp

            records.append(row)

        except Exception as e:
            print(f"Could not read {fp}: {e}")

    df = pd.DataFrame(records)

    if df.empty:
        raise ValueError(f"No valid solution JSON files found in {results_dir}")

    return df


def load_one_state_wastage_json(
    state,
    SCENARIOS,
    batt_capex=150,
    program_scenario_label="flex",
    baseline_scenario_label="flex",
    adoption_filter="Mid",
    time_step_h=1.0,
):
    df_list = []

    needed_labels = sorted(set([
        baseline_scenario_label,
        program_scenario_label,
    ]))

    for scenario_label in needed_labels:
        results_dir = build_one_results_dir(
            state=state,
            adoption_filter=adoption_filter,
            scenario_label=scenario_label,
        )

        if not os.path.exists(results_dir):
            print(f"Skipping missing folder: {results_dir}")
            continue

        temp = load_wastage_from_results_json(
            results_dir=results_dir,
            SCENARIOS=SCENARIOS,
            scenario_label=scenario_label,
            adoption_filter=adoption_filter,
            time_step_h=time_step_h,
        )

        df_list.append(temp)

    if len(df_list) == 0:
        raise ValueError(f"No wastage data found for {state}")

    df = pd.concat(df_list, ignore_index=True)
    df["state"] = state

    df = df[df["batt_capex_num"] == batt_capex].copy()
    df = df[df["adoption"] == adoption_filter.capitalize()].copy()

    base = df[
        (df["group"] == "Base only")
        & (df["scenario_label"] == baseline_scenario_label)
    ].copy()

    programs = df[
        (df["group"].isin(["V1G", "V2G", "V1G+V2G"]))
        & (df["scenario_label"] == program_scenario_label)
    ].copy()

    out = pd.concat([base, programs], ignore_index=True)

    return out


# ============================================================
# Plot
# ============================================================

def plot_rps_reached_by_wastage_with_adoption_band(
    states=("CA", "NY", "TX"),
    state_names=None,
    SCENARIOS=None,
    batt_capex=150,
    program_scenario_label="flex",
    baseline_scenario_label="flex",
    main_adoption="Mid",
    band_adoptions=("Slow", "Fast"),
    rps_order_by_state=None,
    selected_participation=("30%", "50%"),
    save_path=None,
    time_step_h=1.0,
):
    if SCENARIOS is None:
        raise ValueError("You must pass SCENARIOS.")

    if state_names is None:
        state_names = {"CA": "California", "NY": "New York", "TX": "Texas"}

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    group_order = ["V1G", "V2G", "V1G+V2G"]

    group_colors = {
        "V1G": "#4E79A7",
        "V2G": "#59A14F",
        "V1G+V2G": "#E15759",
    }

    marker_map = {
        "30%": "s",
        "50%": "^",
    }

    # -----------------------------
    # Load Slow, Mid, Fast
    # -----------------------------
    all_data = []

    all_adoptions = list(dict.fromkeys(
        list(band_adoptions) + [main_adoption]
    ))

    for adoption in all_adoptions:
        for state in states:
            base_label = get_label_for_state(baseline_scenario_label, state)
            program_label = get_label_for_state(program_scenario_label, state)

            temp = load_one_state_wastage_json(
                state=state,
                SCENARIOS=SCENARIOS,
                batt_capex=batt_capex,
                program_scenario_label=program_label,
                baseline_scenario_label=base_label,
                adoption_filter=adoption,
                time_step_h=time_step_h,
            )

            all_data.append(temp)

    df = pd.concat(all_data, ignore_index=True)

    df = pd.concat(
        [
            df[
                (df["state"] == state)
                & (df["rps_percent"].isin(rps_order_by_state[state]))
            ]
            for state in states
        ],
        ignore_index=True,
    )

    df = df[
        (df["group"] == "Base only")
        | (df["participation"].isin(selected_participation))
    ].copy()

    # -----------------------------
    # Figure
    # -----------------------------
    fig, axes = plt.subplots(
        1,
        len(states),
        figsize=(16, 5),
        sharey=False,
        facecolor="white",
    )

    if len(states) == 1:
        axes = [axes]

    for ax, state in zip(axes, states):
        state_rps_order = rps_order_by_state[state]
        sub_state = df[df["state"] == state].copy()

        ax.set_facecolor("white")

        # -----------------------------
        # Baseline horizontal bands/lines
        # -----------------------------
        base = sub_state[
            (sub_state["group"] == "Base only")
            & (sub_state["rps_percent"].isin(state_rps_order))
        ].copy()

        for rps in state_rps_order:
            base_rps = base[base["rps_percent"] == rps]

            base_mid = base_rps[
                base_rps["adoption"] == main_adoption.capitalize()
            ]

            base_band = base_rps[
                base_rps["adoption"].isin([a.capitalize() for a in band_adoptions])
            ]

            if base_band.empty and base_mid.empty:
                continue

            if not base_band.empty:
                y_min = base_band["wastage_gwh"].min()
                y_max = base_band["wastage_gwh"].max()

                ax.axhspan(
                    y_min,
                    y_max,
                    color="gray",
                    alpha=0.10,
                    zorder=0,
                )

            if not base_mid.empty:
                base_wastage = base_mid["wastage_gwh"].iloc[0]

                ax.axhline(
                    y=base_wastage,
                    color="gray",
                    linestyle="--",
                    linewidth=1.2,
                    alpha=0.75,
                    zorder=1,
                )

                ax.text(
                    rps,
                    base_wastage,
                    f"Base {rps}%",
                    ha="center",
                    va="bottom",
                    fontsize=8,
                    color="dimgray",
                )

        # -----------------------------
        # Program lines and shaded bands
        # -----------------------------
        for group in group_order:
            for part in selected_participation:
                sub_gp = sub_state[
                    (sub_state["group"] == group)
                    & (sub_state["participation"] == part)
                ].copy()

                if sub_gp.empty:
                    continue

                main = sub_gp[
                    sub_gp["adoption"] == main_adoption.capitalize()
                ].sort_values("rps_percent")

                band = sub_gp[
                    sub_gp["adoption"].isin([a.capitalize() for a in band_adoptions])
                ]

                slow = sub_gp[
                    sub_gp["adoption"] == "Slow"
                ][["rps_percent", "wastage_gwh"]].rename(
                    columns={"wastage_gwh": "slow_wastage"}
                )

                fast = sub_gp[
                    sub_gp["adoption"] == "Fast"
                ][["rps_percent", "wastage_gwh"]].rename(
                    columns={"wastage_gwh": "fast_wastage"}
                )

                band_summary = slow.merge(
                    fast,
                    on="rps_percent",
                    how="inner",
                )

                band_summary = band_summary[
                    band_summary["rps_percent"].isin(state_rps_order)
                ].sort_values("rps_percent")



                if not band_summary.empty:
                    ax.fill_between(
                        band_summary["rps_percent"],
                        band_summary["slow_wastage"],
                        band_summary["fast_wastage"],
                        color=group_colors[group],
                        alpha=0.15,
                        linewidth=0,
                        zorder=2,
                    )

                    # optional: draw the actual slow/fast edges
                    ax.plot(
                        band_summary["rps_percent"],
                        band_summary["slow_wastage"],
                        color=group_colors[group],
                        linestyle=":",
                        linewidth=1.0,
                        alpha=0.7,
                        zorder=2,
                    )

                    ax.plot(
                        band_summary["rps_percent"],
                        band_summary["fast_wastage"],
                        color=group_colors[group],
                        linestyle=":",
                        linewidth=1.0,
                        alpha=0.7,
                        zorder=2,
                    )

                if not main.empty:
                    main = main[main["rps_percent"].isin(state_rps_order)]

                    ax.plot(
                        main["rps_percent"],
                        main["wastage_gwh"],
                        color=group_colors[group],
                        marker=marker_map[part],
                        markerfacecolor=group_colors[group],
                        markeredgecolor="black",
                        linewidth=2.0,
                        markersize=7,
                        label=f"{group}, {part}",
                        zorder=3,
                    )

        # -----------------------------
        # Formatting
        # -----------------------------
        ax.set_title(
            state_names.get(state, state),
            fontsize=13,
            color="black",
        )

        ax.set_xlabel("RPS target (%)", fontsize=11, color="black")
        ax.set_ylabel("Curtailment (GWh)", fontsize=11, color="black")

        ax.set_xticks(state_rps_order)
        ax.set_xlim(min(state_rps_order) - 3, max(state_rps_order) + 5)

        ax.tick_params(axis="both", colors="black", labelcolor="black")
        ax.grid(True, which="both", linestyle="--", alpha=0.3, color="gray")

        for spine in ax.spines.values():
            spine.set_color("black")
            spine.set_linewidth(1.0)

    # -----------------------------
    # Legend
    # -----------------------------
    strategy_handles = [
        Line2D(
            [0], [0],
            color=group_colors["V1G"],
            marker="o",
            markerfacecolor=group_colors["V1G"],
            markeredgecolor="black",
            lw=2,
            label="V1G",
        ),
        Line2D(
            [0], [0],
            color=group_colors["V2G"],
            marker="o",
            markerfacecolor=group_colors["V2G"],
            markeredgecolor="black",
            lw=2,
            label="V2G",
        ),
        Line2D(
            [0], [0],
            color=group_colors["V1G+V2G"],
            marker="o",
            markerfacecolor=group_colors["V1G+V2G"],
            markeredgecolor="black",
            lw=2,
            label="V1G+V2G",
        ),
        Line2D(
            [0], [0],
            color="gray",
            linestyle="--",
            lw=1.5,
            label="Base, Mid",
        ),
    ]

    participation_handles = [
        Line2D(
            [0], [0],
            color="black",
            marker=marker_map["30%"],
            markersize=8,
            markerfacecolor="gray",
            markeredgecolor="black",
            lw=0,
            label="30% participation",
        ),
        Line2D(
            [0], [0],
            color="black",
            marker=marker_map["50%"],
            markersize=8,
            markerfacecolor="black",
            markeredgecolor="black",
            lw=0,
            label="50% participation",
        ),
        Line2D(
            [0], [0],
            color="gray",
            lw=8,
            alpha=0.20,
            label="Slow to Fast range",
        ),
    ]

    legend = fig.legend(
        handles=strategy_handles + participation_handles,
        loc="lower center",
        ncol=7,
        frameon=True,
        bbox_to_anchor=(0.5, -0.04),
        fontsize=10,
    )

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    for text in legend.get_texts():
        text.set_color("black")

    fig.suptitle(
        f"Curtailment under different EV charging strategies\n"
        f"Battery capex = ${batt_capex}/kWh, Mid line with Slow to Fast range",
        fontsize=15,
        y=1.03,
        color="black",
    )

    fig.subplots_adjust(bottom=0.20, top=0.88)

    if save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)

        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white",
        )

    return df, fig, axes

df_wastage_band, fig, axes = plot_rps_reached_by_wastage_with_adoption_band(
    states=("CA", "NY", "TX"),
    SCENARIOS=SCENARIOS,
    batt_capex=250,
    program_scenario_label={
        "CA": "flex",
        "NY": "flex",
        "TX": "midnight",
    },
    baseline_scenario_label={
        "CA": "midnight",
        "NY": "midnight",
        "TX": "arrive",
    },
    main_adoption="Mid",
    band_adoptions=("Slow", "Fast"),
    rps_order_by_state={
        "CA": [50, 60, 70],
        "NY": [60, 70, 80],
        "TX": [50, 60, 70],
    },
    selected_participation=("30%", "50%"),
    save_path="figures/wastage_mid_line_slow_fast_band_bcapex250.png",
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ev_adoption = {
    "CA": {
        2025: {"slow": 2.3e6, "mid": 2.6e6, "fast": 3.4e6},
        2030: {"slow": 5.1e6, "mid": 6.1e6, "fast": 7.2e6},
        2035: {"slow": 11.9e6, "mid": 12.6e6, "fast": 14.2e6},
    },
    "TX": {
        2025: {"slow": 0.46e6, "mid": 0.65e6, "fast": 0.88e6},
        2030: {"slow": 1.50e6, "mid": 2.20e6, "fast": 3.50e6},
        2035: {"slow": 4.50e6, "mid": 7.20e6, "fast": 10.40e6},
    },
    "NY": {
        2025: {"slow": 0.39e6, "mid": 0.53e6, "fast": 0.70e6},
        2030: {"slow": 1.59e6, "mid": 2.3e6, "fast": 3.1e6},
        2035: {"slow": 4.0e6, "mid": 6.1e6, "fast": 9.2e6},
    }
}


def plot_ev_adoption_with_annotation(ev_adoption, save_path=None):

    # -----------------------------
    # Convert to DataFrame
    # -----------------------------
    rows = []
    for state, years in ev_adoption.items():
        for year, scenarios in years.items():
            slow = scenarios["slow"]
            mid = scenarios["mid"]
            fast = scenarios["fast"]

            rows.append({
                "State": state,
                "Year": year,
                "Slow": slow / 1e6,
                "Mid": mid / 1e6,
                "Fast": fast / 1e6,
                "Lower Error": (mid - slow) / 1e6,
                "Upper Error": (fast - mid) / 1e6,
            })

    df = pd.DataFrame(rows)

    # -----------------------------
    # Plot
    # -----------------------------
    sns.set_theme(style="whitegrid", font_scale=1.2)

    fig, ax = plt.subplots(figsize=(10, 6))

    palette = {
        "CA": "#1f77b4",
        "TX": "#ff7f0e",
        "NY": "#2ca02c",
    }

    for state in df["State"].unique():
        state_df = df[df["State"] == state].sort_values("Year")

        ax.errorbar(
            state_df["Year"],
            state_df["Mid"],
            yerr=[state_df["Lower Error"], state_df["Upper Error"]],
            marker="o",
            linewidth=2.5,
            capsize=6,
            elinewidth=1.8,
            label=state,
            color=palette[state],
        )

    # -----------------------------
    # Annotation (ONLY CA 2035)
    # -----------------------------
    # -----------------------------
    # Annotation (ONLY CA 2035)
    # -----------------------------
    ca_2035 = df[(df["State"] == "CA") & (df["Year"] == 2035)].iloc[0]

    x = ca_2035["Year"]
    y_mid = ca_2035["Mid"]
    y_slow = ca_2035["Slow"]
    y_fast = ca_2035["Fast"]

    # Vertical line showing full range
    ax.vlines(
        x=x,
        ymin=y_slow,
        ymax=y_fast,
        colors="gray",
        linestyles="dashed",
        linewidth=1.5,
        alpha=0.7
    )

    # FAST annotation (exact anchor)
    ax.annotate(
        "Fast",
        xy=(x, y_fast),                # exact bound
        xytext=(x + 0.7, y_fast + 0.5),
        arrowprops=dict(
            arrowstyle="->",
            lw=1.5,
            color="black"
        ),
        fontsize=17,
        ha="left"
    )

    # SLOW annotation (exact anchor)
    ax.annotate(
        "Slow",
        xy=(x, y_slow),                # exact bound
        xytext=(x + 0.7, y_slow - 0.8),
        arrowprops=dict(
            arrowstyle="->",
            lw=1.5,
            color="black"
        ),
        fontsize=17,
        ha="left"
    )
    # -----------------------------
    # Formatting
    # -----------------------------
    ax.set_title(
        "Projected EV Adoption with Scenario Uncertainty",
        fontsize=19,
        weight="bold",
        pad=15,
    )

    ax.set_xlabel("Year", fontsize=18)
    ax.set_ylabel("Number of EVs (million)", fontsize=18)

    ax.set_xticks(sorted(df["Year"].unique()))

    # Legend (color = state)
    ax.legend(title="State", frameon=True)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    # Turn on minor ticks
    ax.minorticks_on()

    # Minor horizontal grid (light)
    ax.grid(which='minor', axis='y', linestyle=':', linewidth=0.7, alpha=0.5)

    # Major grid (slightly stronger)
    ax.grid(which='major', axis='y', linestyle='--', linewidth=0.9, alpha=0.7)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")

    plt.show()

    return df

df_ev = plot_ev_adoption_with_annotation(
    ev_adoption,
    save_path="ev_adoption_errorbars.png"
)
```

```python
import re

def parse_ev_file_key(key):
    """
    Handles names like:

    CA2026fast_arrivescen0_res_min_delay_temp_gridLoad
    CA2026mid_arrivescen0_res_min_delay_temp_gridLoad
    CA2026slowscen0_res_max_delay_temp_gridLoad
    CA2026midscen1_res_timed_charging_temp_gridLoad
    CA2026slowscen0_res_load_leveling_temp_gridLoad
    """

    # Remove possible duplicated suffix from flattened JSON keys
    key = key.replace(":load", "")

    pattern = (
        r"(?P<state>[A-Z]{2})"
        r"(?P<year>\d{4})"
        r"(?P<adoption>slow|mid|fast)"
        r"(?:_(?P<strategy_before_scen>[A-Za-z]+))?"
        r"scen(?P<scenario>\d+)_"
        r"(?P<result_name>.+?)_temp_gridLoad"
    )

    match = re.search(pattern, key)

    if match is None:
        return None

    strategy_before_scen = match.group("strategy_before_scen")
    result_name = match.group("result_name")

    if strategy_before_scen is not None:
        strategy = strategy_before_scen
    else:
        strategy = result_name.replace("res_", "")

    strategy_map = {
        "arrive": "Arrive",
        "even": "Load leveling",
        "max_delay": "Max delay",
        "timed_charging": "Timed charging",
        "load_leveling": "Load leveling",
        "min_delay": "Min delay",
    }

    strategy_clean = strategy_map.get(strategy, strategy)

    return {
        "state": match.group("state"),
        "year": int(match.group("year")),
        "adoption": match.group("adoption"),
        "scenario": int(match.group("scenario")),
        "strategy": strategy_clean,
        "raw_strategy": strategy,
    }


def load_ev_charging_json(json_path):
    with open(json_path, "r") as f:
        data = json.load(f)

    rows = []

    for key, value in data["files"].items():
        meta = parse_ev_file_key(key)

        if meta is None:
            print(f"Skipped because name was not parsed: {key}")
            continue

        hourly_kw = value["hourly_total_kw"]

        for hour, kw in enumerate(hourly_kw):
            rows.append({
                "file_key": key,
                "state": meta["state"],
                "year": meta["year"],
                "adoption": meta["adoption"],
                "strategy": meta["strategy"],
                "hour": hour,
                "kw": kw,
                "mw": kw / 1000,
            })

    return pd.DataFrame(rows)

def plot_2030_charging_strategies_with_adoption_shade(
    json_path,
    state="CA",
    year=2030,
    strategies=None,
    aggregate="daily_mean",
    save_path=None,
    figsize=(12, 6),
):
    TITLE_SIZE = 20
    LABEL_SIZE = 18
    TICK_SIZE = 17
    LEGEND_SIZE = 17
    LEGEND_TITLE_SIZE = 17
    df = load_ev_charging_json(json_path)

    df = df[
        (df["state"] == state) &
        (df["year"] == year)
    ].copy()

    if df.empty:
        raise ValueError(f"No data found for {state} in {year}")

    strategy_colors = {
        "Arrive": "#4C72B0",          # blue
        "Max delay": "#DD8452",       # orange
        "Timed charging": "#55A868",  # green
        "Load leveling": "#C44E52",   # red
        "Min delay": "#8172B3",       # purple
        "Flex": "#937860",            # brown
        "Midnight": "#64B5CD",        # cyan
    }

    if strategies is None:
        strategies = [
            s for s in strategy_colors.keys()
            if s in df["strategy"].unique()
        ]

        extra_strategies = [
            s for s in sorted(df["strategy"].unique())
            if s not in strategies
        ]

        strategies = strategies + extra_strategies

    df = df[df["strategy"].isin(strategies)].copy()

    if df.empty:
        raise ValueError("No data found for the selected strategies.")

    if aggregate == "hourly":
        plot_df = df.copy()
        plot_df["x"] = plot_df["hour"]
        x_label = "Hour of year"
        title_time = "Hourly"

    elif aggregate == "daily_mean":
        df["hour_of_day"] = df["hour"] % 24

        plot_df = (
            df.groupby(["strategy", "adoption", "hour_of_day"], as_index=False)["mw"]
            .mean()
        )
        plot_df = plot_df.rename(columns={"hour_of_day": "x"})
        x_label = "Hour of day"
        title_time = "Average daily"

    elif aggregate == "weekly_mean":
        df["hour_of_week"] = df["hour"] % 168

        plot_df = (
            df.groupby(["strategy", "adoption", "hour_of_week"], as_index=False)["mw"]
            .mean()
        )
        plot_df = plot_df.rename(columns={"hour_of_week": "x"})
        x_label = "Hour of week"
        title_time = "Average weekly"

    else:
        raise ValueError("aggregate must be 'hourly', 'daily_mean', or 'weekly_mean'")

    fig, ax = plt.subplots(figsize=figsize)

    fig.patch.set_facecolor("white")
    ax.set_facecolor("white")

    for strategy in strategies:
        temp = plot_df[plot_df["strategy"] == strategy]

        mid = temp[temp["adoption"] == "mid"].sort_values("x")
        slow = temp[temp["adoption"] == "slow"].sort_values("x")
        fast = temp[temp["adoption"] == "fast"].sort_values("x")

        if mid.empty:
            print(f"Skipped {strategy} because mid adoption is missing.")
            continue

        color = strategy_colors.get(strategy, "#333333")

        ax.plot(
            mid["x"],
            mid["mw"],
            linewidth=2.7,
            label=strategy,
            color=color
        )

        if not slow.empty and not fast.empty:
            shade = pd.merge(
                slow[["x", "mw"]],
                fast[["x", "mw"]],
                on="x",
                suffixes=("_slow", "_fast")
            )

            shade["lower"] = shade[["mw_slow", "mw_fast"]].min(axis=1)
            shade["upper"] = shade[["mw_slow", "mw_fast"]].max(axis=1)

            ax.fill_between(
                shade["x"],
                shade["lower"],
                shade["upper"],
                color=color,
                alpha=0.18,
                linewidth=0
            )

    ax.set_title(
        f"{title_time} EV charging demand by strategy in {state}, {year}",
        fontsize=TITLE_SIZE,
        color="black"
    )

    ax.set_xlabel(x_label, fontsize=LABEL_SIZE, color="black")
    ax.set_ylabel("EV charging demand (MW)", fontsize=LABEL_SIZE, color="black")

    ax.tick_params(axis="both", labelsize=TICK_SIZE, colors="black")

    for spine in ax.spines.values():
        spine.set_color("black")

    ax.grid(True, color="gray", alpha=0.3)

    legend = ax.legend(
    title="Charging strategy",
    fontsize=LEGEND_SIZE,
    title_fontsize=LEGEND_TITLE_SIZE,
    frameon=True,
    facecolor="white",
    edgecolor="black"
    )

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    legend.get_title().set_color("black")

    for text in legend.get_texts():
        text.set_color("black")

    if aggregate == "daily_mean":
        ax.set_xticks(range(0, 24, 3))

    if aggregate == "weekly_mean":
        ax.set_xticks(range(0, 169, 24))

    plt.tight_layout()

    if save_path is not None:
        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white"
        )

    plt.show()

    return plot_df
```

```python
import os
import re
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
def load_total_costs_from_results(results_dir, scenarios=None, scenario_label=None):
    records = []

    total_files = glob.glob(os.path.join(results_dir, "**", "*_total_cost.csv"), recursive=True)
    summary_files = glob.glob(os.path.join(results_dir, "**", "*_summary.csv"), recursive=True)
    files = sorted(total_files + summary_files)

    print(f"Found {len(files)} candidate files in {results_dir}")

    seen = set()

    for fp in files:
        try:
            temp = pd.read_csv(fp)
            if temp.empty:
                continue

            row = temp.iloc[0].to_dict()

            fname = os.path.basename(fp)
            folder = os.path.basename(os.path.dirname(fp))

            scenario_id = int(fname.split("_")[0][1:])

            # prefer total_cost over summary
            if scenario_id in seen and fp.endswith("_summary.csv"):
                continue

            cfg = {}
            cfg["scenario_id"] = scenario_id
            cfg["scenario_tag"] = row.get("scenario_tag", "")
            cfg["objective_value"] = float(row["objective_value"])
            cfg["file_path"] = fp

            base_dir_name = os.path.basename(os.path.normpath(results_dir)).lower()

            if "_slow_" in f"_{base_dir_name}_":
                cfg["adoption"] = "Slow"
            elif "_fast_" in f"_{base_dir_name}_":
                cfg["adoption"] = "Fast"
            elif "_mid_" in f"_{base_dir_name}_":
                cfg["adoption"] = "Mid"
            else:
                cfg["adoption"] = "Unknown"

            # NEW: Add scenario label from directory
            if scenario_label is None:
                # Extract label from directory name (e.g., "flex", "midnight")
                parts = base_dir_name.split("_")
                cfg["scenario_label"] = parts[-1] if len(parts) > 0 else "default"
            else:
                cfg["scenario_label"] = scenario_label

            # Extract v1g_share and v2g_share from folder name
            m_v1g = re.search(r"v1g(\d+)", folder, re.IGNORECASE)
            m_v2g = re.search(r"v2g(\d+)", folder, re.IGNORECASE)

            v1g_val = int(m_v1g.group(1)) if m_v1g else 0
            v2g_val = int(m_v2g.group(1)) if m_v2g else 0

            cfg["v1g_share"] = v1g_val / 100.0
            cfg["v2g_share"] = v2g_val / 100.0

            # Determine participation level (10% or 30%)
            total_participation = v1g_val + v2g_val
            if total_participation == 0:
                cfg["participation"] = "0%"
                cfg["group"] = "Base only"
            elif total_participation == 10:
                cfg["participation"] = "10%"
            elif total_participation == 30:
                cfg["participation"] = "30%"
            elif total_participation == 50:
                cfg["participation"] = "50%"
            else:
                cfg["participation"] = f"{total_participation}%"

            # Determine group
            if v1g_val > 0 and v2g_val > 0:
                cfg["group"] = "V1G+V2G"
            elif v1g_val > 0:
                cfg["group"] = "V1G"
            elif v2g_val > 0:
                cfg["group"] = "V2G"
            else:
                cfg["group"] = "Base only"

            m = re.search(r"rps(\d+)", folder)
            cfg["rps_ratio"] = int(m.group(1)) / 100 if m else None

            m = re.search(r"bcapex(\d+)", folder)
            if m:
                batt = int(m.group(1))
                cfg["batt_capex_label"] = f"${batt}/kWh"
            else:
                cfg["batt_capex_label"] = None

            records = [r for r in records if r["scenario_id"] != scenario_id]
            records.append(cfg)
            seen.add(scenario_id)

        except Exception as e:
            print(f"Could not read {fp}: {e}")

    df = pd.DataFrame(records)

    if df.empty:
        raise ValueError(f"No valid total cost or summary files found in: {results_dir}")

    df = df.sort_values("scenario_id").reset_index(drop=True)
    print("Available scenario IDs:", sorted(df["scenario_id"].unique().tolist()))
    print("Participation levels found:", df["participation"].dropna().unique())

    return df

def load_cost_components_from_results(results_dir, scenario_label=None):
    records = []

    # Use total_cost/summary reader as metadata source
    meta_df = load_total_costs_from_results(results_dir, scenarios=None, scenario_label=scenario_label)
    meta_map = {
        int(row["scenario_id"]): row
        for _, row in meta_df.iterrows()
    }

    component_files = glob.glob(
        os.path.join(results_dir, "**", "*_cost_components.csv"),
        recursive=True
    )

    print(f"Found {len(component_files)} cost component files in {results_dir}")

    for fp in sorted(component_files):
        try:
            temp = pd.read_csv(fp)
            if temp.empty:
                continue

            row = temp.iloc[0].to_dict()
            fname = os.path.basename(fp)

            scenario_id = int(fname.split("_")[0][1:])

            if scenario_id not in meta_map:
                print(f"Could not find metadata match for cost component file: {fp}")
                continue

            meta = meta_map[scenario_id]

            cfg = {}
            cfg["scenario_id"] = scenario_id
            cfg["scenario_tag"] = meta.get("scenario_tag", "")
            cfg["objective_value"] = float(row.get("objective_value", meta.get("objective_value", 0.0)))

            cfg["capex_total"] = float(row.get("capex_total", 0.0))
            cfg["asset_opex_total"] = float(row.get("asset_opex_total", 0.0))
            cfg["line_opex_total"] = float(row.get("line_opex_total", 0.0))
            cfg["fixed_total"] = float(row.get("fixed_total", 0.0))
            cfg["penalty_total"] = float(row.get("penalty_total", 0.0))
            cfg["other_gap"] = float(row.get("other_gap", 0.0))
            cfg["file_path"] = fp

            # Bring all plotting metadata from total_cost/summary loader
            cfg["adoption"] = meta.get("adoption", "Unknown")
            cfg["scenario_label"] = meta.get("scenario_label", scenario_label if scenario_label is not None else "default")
            cfg["v1g_share"] = float(meta.get("v1g_share", 0.0))
            cfg["v2g_share"] = float(meta.get("v2g_share", 0.0))
            cfg["participation"] = meta.get("participation", "0%")
            cfg["group"] = meta.get("group", "Base only")
            cfg["rps_ratio"] = meta.get("rps_ratio", None)

            batt_label = meta.get("batt_capex_label", None)
            if batt_label is not None:
                m = re.search(r"(\d+)", str(batt_label))
                cfg["batt_capex_num"] = float(m.group(1)) if m else None
            else:
                cfg["batt_capex_num"] = None

            records.append(cfg)

        except Exception as e:
            print(f"Could not read {fp}: {e}")

    df = pd.DataFrame(records)

    if not df.empty:
        print("Cost component scenario IDs:", sorted(df["scenario_id"].unique().tolist()))
        if "batt_capex_num" in df.columns:
            print("Cost component battery capex values:", sorted(df["batt_capex_num"].dropna().unique().tolist()))

    return df


component_colors = {
    "capex": "#7B6D8D",
    "asset_opex": "#59A14F",
    "line_opex": "#E15759",
    "other": "#C7C7C7",
}


def plot_base_cost_by_state_rps_bcapex(
    results_dirs_by_state,
    batt_capex=150,
    states=("CA", "NY", "TX"),
    rps_order_by_state=None,
    baseline_order=("midnight", "arrive", "flex"),
    save_path=None,
    figsize=(18, 6),
    xlim_by_state=None,
):
    """
    Plot only Base scenarios for three states.

    Layout
    ------
    1 x 3 panels:
        CA | NY | TX

    Each panel includes:
        RPS sections
        Base charging scenarios within each RPS
        Mid adoption as stacked horizontal bar
        Slow and Fast adoption as horizontal error range

    Parameters
    ----------
    results_dirs_by_state : dict
        Example:
        {
            "CA": {
                "midnight": [...],
                "arrive": [...],
                "flex": [...]
            },
            "NY": {
                "midnight": [...],
                "arrive": [...],
                "flex": [...]
            },
            "TX": {
                "midnight": [...],
                "arrive": [...],
                "flex": [...]
            }
        }

    batt_capex : int
        Battery capex value, for example 150 or 250.

    states : tuple
        States to plot.

    rps_order_by_state : dict or None
        Example:
        {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70]
        }

    baseline_order : tuple
        Order of base charging scenarios.

    save_path : str or None
        Output path.
    """

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    fig, axes = plt.subplots(
        1,
        len(states),
        figsize=figsize,
        sharex=False,
        sharey=False
    )

    if len(states) == 1:
        axes = [axes]

    fig.patch.set_facecolor("white")

    bar_height = 0.55
    scenario_gap = 0.18
    rps_gap = 0.70

    all_plot_data = []

    def draw_stacked_bar(ax, y, row):
        capex = row["capex_total_mil"]
        asset_opex = row["asset_opex_total_mil"]
        line_opex = row["line_opex_total_mil"]
        other = row["other_total_mil"]

        left = 0.0

        ax.barh(
            y, capex, left=left, height=bar_height,
            color=component_colors["capex"],
            edgecolor="black", linewidth=0.5
        )
        left += capex

        ax.barh(
            y, asset_opex, left=left, height=bar_height,
            color=component_colors["asset_opex"],
            edgecolor="black", linewidth=0.5
        )
        left += asset_opex

        ax.barh(
            y, line_opex, left=left, height=bar_height,
            color=component_colors["line_opex"],
            edgecolor="black", linewidth=0.5
        )
        left += line_opex

        ax.barh(
            y, other, left=left, height=bar_height,
            color=component_colors["other"],
            edgecolor="black", linewidth=0.5,
            hatch="///"
        )

    for ax, state in zip(axes, states):
        ax.set_facecolor("white")

        if state not in results_dirs_by_state:
            raise ValueError(f"{state} is missing from results_dirs_by_state")

        df_list = []

        for scenario_label, dirs in results_dirs_by_state[state].items():
            for rd in dirs:
                temp = load_cost_components_from_results(
                    rd,
                    scenario_label=scenario_label
                )

                if temp.empty:
                    continue

                df_list.append(temp)

        if len(df_list) == 0:
            raise ValueError(f"No cost component files found for {state}")

        df = pd.concat(df_list, ignore_index=True)

        # Keep only selected battery capex
        df = df[df["batt_capex_num"] == batt_capex].copy()

        # Keep only Base scenario
        df = df[df["group"] == "Base only"].copy()

        if df.empty:
            raise ValueError(
                f"No Base only rows found for {state} with battery capex ${batt_capex}/kWh"
            )

        df["rps_percent"] = (df["rps_ratio"] * 100).round().astype(int)

        cost_cols = [
            "capex_total",
            "asset_opex_total",
            "line_opex_total",
            "fixed_total",
            "penalty_total",
            "other_gap",
            "objective_value",
        ]

        for col in cost_cols:
            df[col + "_mil"] = df[col] / 1e6

        df["other_total_mil"] = (
            df["fixed_total_mil"]
            + df["penalty_total_mil"]
            + df["other_gap_mil"]
        )

        df["objective_value_mil"] = df["objective_value"] / 1e6

        all_plot_data.append(df.assign(state=state))

        y_positions = []
        y_labels = []
        rps_section_centers = []

        current_y = 0

        rps_order = rps_order_by_state.get(state, sorted(df["rps_percent"].unique()))

        for rps in rps_order:
            sub_rps = df[df["rps_percent"] == rps].copy()

            if sub_rps.empty:
                continue

            section_start_y = current_y

            for scenario_label in baseline_order:
                sub_scenario = sub_rps[sub_rps["scenario_label"] == scenario_label].copy()

                mid = sub_scenario[sub_scenario["adoption"] == "Mid"]
                slow = sub_scenario[sub_scenario["adoption"] == "Slow"]
                fast = sub_scenario[sub_scenario["adoption"] == "Fast"]

                if mid.empty:
                    continue

                mid_row = mid.iloc[0]
                mid_val = mid_row["objective_value_mil"]

                slow_val = slow["objective_value_mil"].iloc[0] if not slow.empty else mid_val
                fast_val = fast["objective_value_mil"].iloc[0] if not fast.empty else mid_val

                low_val = min(slow_val, mid_val, fast_val)
                high_val = max(slow_val, mid_val, fast_val)

                left_err = mid_val - low_val
                right_err = high_val - mid_val

                draw_stacked_bar(ax, current_y, mid_row)

                ax.errorbar(
                    x=mid_val,
                    y=current_y,
                    xerr=np.array([[left_err], [right_err]]),
                    fmt="none",
                    ecolor="black",
                    elinewidth=1.5,
                    capsize=4,
                    capthick=1.5,
                    zorder=4
                )

                ax.text(
                    0.01 * ax.get_xlim()[1] if ax.get_xlim()[1] > 1 else 0.5,
                    current_y,
                    scenario_label.upper(),
                    va="center",
                    ha="left",
                    fontsize=9,
                    fontweight="bold",
                    color="white",
                    zorder=5
                )

                y_positions.append(current_y)
                y_labels.append(scenario_label.capitalize())

                current_y += bar_height + scenario_gap

            section_end_y = current_y - scenario_gap
            section_center = (section_start_y + section_end_y) / 2
            rps_section_centers.append((rps, section_center))

            current_y += rps_gap

            # Divider line between RPS sections
            ax.axhline(
                current_y - rps_gap / 2,
                color="black",
                linewidth=0.6,
                alpha=0.25
            )

        ax.set_yticks(y_positions)
        ax.set_yticklabels(y_labels, fontsize=10, color="black")
        ax.invert_yaxis()

        ax.set_title(
            state,
            fontsize=18,
            fontweight="bold",
            color="black"
        )

        ax.set_xlabel(
            "Total system cost (Million $)",
            fontsize=16,
            color="black"
        )

        ax.tick_params(axis="both", labelsize=16, colors="black")
        ax.grid(True, axis="x", linestyle="--", alpha=0.3)

        for spine in ax.spines.values():
            spine.set_color("black")

        # Add RPS labels on the right side of each panel
        x_max = df["objective_value_mil"].max() * 1.18

        if xlim_by_state is not None and state in xlim_by_state:
            ax.set_xlim(xlim_by_state[state])
            x_text = xlim_by_state[state][1] * 0.97
        else:
            ax.set_xlim(0, x_max)
            x_text = x_max * 0.99

        for rps, y_center in rps_section_centers:
            ax.text(
                x_text,
                y_center,
                f"RPS {rps}%",
                va="center",
                ha="right",
                fontsize=9,
                fontweight="bold",
                color="black",
                bbox=dict(
                    facecolor="white",
                    edgecolor="black",
                    boxstyle="round,pad=0.25",
                    linewidth=0.6
                )
            )

    axes[0].set_ylabel(
        "Base charging strategy",
        fontsize=18,
        color="black"
    )

    fig.suptitle(
        f"Base EV charging strategies affect system cost across RPS targets\n"
        f"Battery capex = ${int(batt_capex)}/kWh",
        fontsize=18,
        color="black",
        y=0.98
    )

    legend_handles = [
        Patch(facecolor=component_colors["capex"], edgecolor="black", label="Asset CAPEX"),
        Patch(facecolor=component_colors["asset_opex"], edgecolor="black", label="Asset OPEX"),
        Patch(facecolor=component_colors["line_opex"], edgecolor="black", label="Line OPEX"),
        Patch(facecolor=component_colors["other"], edgecolor="black", hatch="///", label="Fixed / Penalty / Other"),
    ]

    legend = fig.legend(
        handles=legend_handles,
        loc="lower center",
        ncol=4,
        frameon=True,
        fontsize=16,
        bbox_to_anchor=(0.5, -0.02)
    )

    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    for text in legend.get_texts():
        text.set_color("black")

    plt.tight_layout(rect=[0.02, 0.07, 1, 0.92])

    if save_path is not None:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)

        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white"
        )

    return pd.concat(all_plot_data, ignore_index=True), fig, axes

results_dirs_by_state = {}

for state in ["CA", "NY", "TX"]:
    results_dirs_by_state[state] = {
        "midnight": [
            f"scenario_results_2030_slow_even_{state}_midnight",
            f"scenario_results_2030_mid_even_{state}_midnight",
            f"scenario_results_2030_fast_even_{state}_midnight",
        ],
        "arrive": [
            f"scenario_results_2030_slow_even_{state}_arrive",
            f"scenario_results_2030_mid_even_{state}_arrive",
            f"scenario_results_2030_fast_even_{state}_arrive",
        ],
        "flex": [
            f"scenario_results_2030_slow_even_{state}_flex",
            f"scenario_results_2030_mid_even_{state}_flex",
            f"scenario_results_2030_fast_even_{state}_flex",
        ],
        "delay": [
            f"scenario_results_2030_slow_even_{state}_delay",
            f"scenario_results_2030_mid_even_{state}_delay",
            f"scenario_results_2030_fast_even_{state}_delay",
        ],
    }


df_base_cost, fig, axes = plot_base_cost_by_state_rps_bcapex(
    results_dirs_by_state=results_dirs_by_state,
    batt_capex=150,
    states=("CA", "NY", "TX"),
    rps_order_by_state={
        "CA": [50, 60, 70],
        "NY": [60, 70, 80],
        "TX": [50, 60, 70],
    },
    baseline_order=("midnight", "arrive", "delay", "flex"),
    save_path="figures/base_cost_by_state_rps_bcapex150.png",
)
```

```python
import os
import re
import glob
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch


def get_label_for_state(label_input, state):
    if isinstance(label_input, dict):
        return label_input[state]
    return label_input


def parse_scenario_metadata(folder_name):
    meta = {}

    m = re.search(r"^s(\d+)_", folder_name)
    meta["scenario_id"] = int(m.group(1)) if m else None

    m_v1g = re.search(r"v1g(\d+)", folder_name, re.IGNORECASE)
    m_v2g = re.search(r"v2g(\d+)", folder_name, re.IGNORECASE)

    v1g_val = int(m_v1g.group(1)) if m_v1g else 0
    v2g_val = int(m_v2g.group(1)) if m_v2g else 0

    total_participation = v1g_val + v2g_val

    if total_participation == 0:
        meta["participation"] = "0%"
        meta["group"] = "Base only"
    elif total_participation in [10, 30, 50]:
        meta["participation"] = f"{total_participation}%"
    else:
        meta["participation"] = f"{total_participation}%"

    if "_Base_only_" in folder_name:
        meta["group"] = "Base only"
    elif v1g_val > 0 and v2g_val > 0:
        meta["group"] = "V1G+V2G"
    elif v1g_val > 0:
        meta["group"] = "V1G"
    elif v2g_val > 0:
        meta["group"] = "V2G"
    else:
        meta["group"] = "Base only"

    m = re.search(r"rps(\d+)", folder_name)
    meta["rps"] = int(m.group(1)) if m else None

    m = re.search(r"bcapex(\d+)", folder_name)
    meta["batt_capex"] = int(m.group(1)) if m else None

    return meta


def build_line_capex_cost_lookup(graph_json_path):
    with open(graph_json_path, "r") as f:
        graph = json.load(f)

    lookup = {}
    for edge in graph.get("edges", []):
        for line_id, line_data in edge.get("lines", {}).items():
            lookup[line_id] = line_data.get("capex_cost", 0.0)

    return lookup


def line_expansion_cost_from_solution_json(solution_json_path, line_capex_lookup, amortization):
    with open(solution_json_path, "r") as f:
        sol = json.load(f)

    total_expansion_mw = 0.0
    total_expansion_cost_mil = 0.0

    for edge in sol.get("edges", []):
        for line_id, line_data in edge.get("lines", {}).items():
            capex_raw = line_data.get("capex", [0.0])

            if isinstance(capex_raw, list):
                capex_w = capex_raw[0] if capex_raw else 0.0
            else:
                capex_w = capex_raw

            capex_cost = line_capex_lookup.get(line_id, 0.0)

            total_expansion_mw += capex_w / 1e6
            total_expansion_cost_mil += (capex_w * capex_cost * amortization) / 1e6

    return total_expansion_mw, total_expansion_cost_mil


def line_opex_from_cost_components(folder, verbose=False):
    files = glob.glob(os.path.join(folder, "*cost_components.csv"))

    if verbose:
        print("\nFolder:", folder)
        print("Cost component files found:", files)

    if not files:
        return 0.0

    df = pd.read_csv(files[0])

    if verbose:
        print(df.columns.tolist())
        print(df.head())

    if df.empty:
        return 0.0

    if "line_opex_total" not in df.columns:
        raise ValueError(
            f"'line_opex_total' not found in {files[0]}. "
            f"Columns are: {df.columns.tolist()}"
        )

    return float(df.iloc[0]["line_opex_total"]) / 1e6

def collect_line_cost_for_state(
    state,
    adoption,
    graph_json_path,
    batt_capex_target=150,
    baseline_scenario_label="arrive",
    program_scenario_label="flex",
    selected_groups=("V1G", "V2G", "V1G+V2G"),
    selected_participation=("30%", "50%"),
    amortization=7 / 365,
):
    line_capex_lookup = build_line_capex_cost_lookup(graph_json_path)

    baseline_dir = f"scenario_results_2030_{adoption}_even_{state}_{baseline_scenario_label}"
    program_dir = f"scenario_results_2030_{adoption}_even_{state}_{program_scenario_label}"

    # Read line OPEX from cost_components.csv using your working function
    base_cost_df = load_cost_components_from_results(
        baseline_dir,
        scenario_label=baseline_scenario_label
    )

    program_cost_df = load_cost_components_from_results(
        program_dir,
        scenario_label=program_scenario_label
    )

    cost_df = pd.concat([base_cost_df, program_cost_df], ignore_index=True)

    if cost_df.empty:
        raise ValueError(f"No cost component data found for {state}")

    cost_df = cost_df[cost_df["batt_capex_num"] == batt_capex_target].copy()
    cost_df["rps"] = (cost_df["rps_ratio"] * 100).round().astype(int)
    cost_df["line_opex_mil"] = cost_df["line_opex_total"] / 1e6

    rows = []

    def read_solution_folder(folder, scenario_label):
        folder_name = os.path.basename(folder)
        meta = parse_scenario_metadata(folder_name)

        if meta["batt_capex"] != batt_capex_target:
            return None

        solution_files = glob.glob(os.path.join(folder, "*_solution.json"))
        if not solution_files:
            return None

        expansion_mw, expansion_cost_mil = line_expansion_cost_from_solution_json(
            solution_json_path=solution_files[0],
            line_capex_lookup=line_capex_lookup,
            amortization=amortization,
        )

        # Match line OPEX from cost_components.csv
        match = cost_df[
            (cost_df["scenario_id"] == meta["scenario_id"])
            & (cost_df["scenario_label"] == scenario_label)
        ]

        if match.empty:
            line_opex_mil = 0.0
        else:
            line_opex_mil = match["line_opex_mil"].iloc[0]

        return {
            **meta,
            "state": state,
            "scenario_label": scenario_label,
            "line_expansion_mw": expansion_mw,
            "line_expansion_cost_mil": expansion_cost_mil,
            "line_opex_mil": line_opex_mil,
            "total_line_cost_mil": expansion_cost_mil + line_opex_mil,
        }

    # Baseline
    for folder in sorted(glob.glob(os.path.join(baseline_dir, "s*"))):
        row = read_solution_folder(folder, baseline_scenario_label)
        if row is None:
            continue
        if row["group"] == "Base only":
            rows.append(row)

    # Program
    for folder in sorted(glob.glob(os.path.join(program_dir, "s*"))):
        row = read_solution_folder(folder, program_scenario_label)
        if row is None:
            continue
        if row["group"] in selected_groups and row["participation"] in selected_participation:
            rows.append(row)

    df = pd.DataFrame(rows)

    base = df[df["group"] == "Base only"].copy()

    base = base.rename(columns={
        "line_expansion_cost_mil": "base_line_expansion_cost_mil",
        "line_opex_mil": "base_line_opex_mil",
        "total_line_cost_mil": "base_total_line_cost_mil",
    })

    base = base[[
        "state",
        "rps",
        "batt_capex",
        "base_line_expansion_cost_mil",
        "base_line_opex_mil",
        "base_total_line_cost_mil",
    ]]

    comp = df[df["group"] != "Base only"].copy()

    merged = comp.merge(
        base,
        on=["state", "rps", "batt_capex"],
        how="left",
    )

    merged["delta_line_expansion_cost_mil"] = (
        merged["line_expansion_cost_mil"]
        - merged["base_line_expansion_cost_mil"]
    )

    merged["delta_line_opex_mil"] = (
        merged["line_opex_mil"]
        - merged["base_line_opex_mil"]
    )

    merged["delta_total_line_cost_mil"] = (
        merged["total_line_cost_mil"]
        - merged["base_total_line_cost_mil"]
    )

    return merged

def plot_line_cost_delta_1x3(
    states=("CA", "NY", "TX"),
    state_names=None,
    graph_json_path_by_state=None,
    adoption="mid",
    batt_capex_target=250,
    baseline_scenario_label_by_state="arrive",
    program_scenario_label_by_state="flex",
    selected_groups=("V1G", "V2G", "V1G+V2G"),
    selected_participation=("50%",),
    rps_order_by_state=None,
    amortization=7 / 365,
    save_path=None,
):
    if state_names is None:
        state_names = {"CA": "California", "NY": "New York", "TX": "Texas"}

    if graph_json_path_by_state is None:
        graph_json_path_by_state = {
            "CA": "Examples/California_new.json",
            "NY": "Examples/New_York.json",
            "TX": "Examples/Texas.json",
        }

    if rps_order_by_state is None:
        rps_order_by_state = {
            "CA": [50, 60, 70],
            "NY": [60, 70, 80],
            "TX": [50, 60, 70],
        }

    component_colors = {
        "expansion": "#7B6D8D",
        "opex": "#E15759",
    }

    def stacked_bar_signed(ax, x, expansion_val, opex_val, width):
        pos_bottom = 0.0
        neg_bottom = 0.0

        for val, label in [
            (expansion_val, "expansion"),
            (opex_val, "opex"),
        ]:
            if abs(val) < 1e-12:
                continue

            if val >= 0:
                bottom = pos_bottom
                pos_bottom += val
            else:
                bottom = neg_bottom
                neg_bottom += val

            ax.bar(
                x,
                val,
                width=width,
                bottom=bottom,
                color=component_colors[label],
                edgecolor="black",
                linewidth=0.6,
                zorder=3,
            )

            # label inside bar component
            if abs(val) > 0.5:
                ax.text(
                    x,
                    bottom + val / 2,
                    "CAPEX" if label == "expansion" else "OPEX",
                    ha="center",
                    va="center",
                    fontsize=12,
                    rotation=90,
                    color="black",
                    zorder=5,
                )

    # -----------------------------
    # Collect data
    # -----------------------------
    df_list = []

    for state in states:
        temp = collect_line_cost_for_state(
            state=state,
            adoption=adoption,
            graph_json_path=get_label_for_state(graph_json_path_by_state, state),
            batt_capex_target=batt_capex_target,
            baseline_scenario_label=get_label_for_state(baseline_scenario_label_by_state, state),
            program_scenario_label=get_label_for_state(program_scenario_label_by_state, state),
            selected_groups=selected_groups,
            selected_participation=selected_participation,
            amortization=amortization,
        )
        df_list.append(temp)

    df = pd.concat(df_list, ignore_index=True)

    df = pd.concat(
        [
            df[
                (df["state"] == state)
                & (df["rps"].isin(rps_order_by_state[state]))
            ]
            for state in states
        ],
        ignore_index=True,
    )

    # -----------------------------
    # Plot
    # -----------------------------
    fig, axes = plt.subplots(
        1,
        len(states),
        figsize=(7.5 * len(states), 7),
        sharey=False,
        facecolor="white",
    )

    if len(states) == 1:
        axes = [axes]

    bar_width = 0.85
    group_gap = .50

    for ax, state in zip(axes, states):
        ax.set_facecolor("white")

        state_df = df[df["state"] == state].copy()
        rps_values = rps_order_by_state[state]

        x_positions = []
        x_labels = []
        rps_centers = []
        separator_positions = []

        current_x = 0

        for rps_idx, rps in enumerate(rps_values):
            local_positions = []

            for group in selected_groups:
                x_positions.append(current_x)
                x_labels.append(group)
                local_positions.append(current_x)

                temp = state_df[
                    (state_df["rps"] == rps)
                    & (state_df["group"] == group)
                    & (state_df["participation"] == selected_participation[0])
                ]

                if temp.empty:
                    expansion_val = 0.0
                    opex_val = 0.0
                else:
                    expansion_val = temp["delta_line_expansion_cost_mil"].sum()
                    opex_val = temp["delta_line_opex_mil"].sum()

                stacked_bar_signed(
                    ax=ax,
                    x=current_x,
                    expansion_val=expansion_val,
                    opex_val=opex_val,
                    width=bar_width,
                )

                current_x += 1

            rps_centers.append(np.mean(local_positions))

            if rps_idx < len(rps_values) - 1:
                separator_positions.append(current_x - 0.5 + group_gap / 2)

            current_x += group_gap

        ax.axhline(0, color="black", linewidth=1.1)

        for xpos in separator_positions:
            ax.axvline(
                xpos,
                color="black",
                linestyle="--",
                linewidth=1.0,
                alpha=0.6,
            )

        ax.set_xticks(x_positions)
        ax.set_xticklabels(
            x_labels,
            fontsize=12,
            # fontweight="bold",
            rotation=25,
            ha="right",
            color="black",
        )

        # Add RPS labels below strategy labels
        ymin, ymax = ax.get_ylim()
        yrange = ymax - ymin if ymax != ymin else 1

        rps_y = ymin - 0.23 * yrange

        for rps, center in zip(rps_values, rps_centers):
            ax.text(
                center,
                rps_y,
                f"RPS {rps}%",
                ha="center",
                va="top",
                fontsize=13,
                fontweight="bold",
                color="black",
                clip_on=False,
            )

        ax.set_title(
            state_names.get(state, state),
            fontsize=18,
            fontweight="bold",
            pad=15,
            color="black",
        )

        # ax.set_xlabel("Charging strategy within each RPS target", fontsize=14)
        ax.set_ylabel(
            "Change in line cost relative to Base ($M)",
            fontsize=15,
            color="black",
        )

        ax.grid(axis="y", linestyle="--", alpha=0.3)
        ax.tick_params(axis="both", labelsize=12, colors="black")

        for spine in ax.spines.values():
            spine.set_color("black")

        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    legend_elements = [
        Patch(
            facecolor=component_colors["expansion"],
            edgecolor="black",
            label="Line expansion CAPEX",
        ),
        Patch(
            facecolor=component_colors["opex"],
            edgecolor="black",
            label="Line OPEX",
        ),
    ]

    legend = fig.legend(
        handles=legend_elements,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.00),
        ncol=2,
        frameon=True,
        fontsize=14,
        title="Line cost component",
    )

    legend.get_title().set_fontsize(15)
    legend.get_title().set_weight("bold")
    legend.get_frame().set_facecolor("white")
    legend.get_frame().set_edgecolor("black")

    fig.suptitle(
        f"Transmission cost change relative to Base\n"
        f"Battery capex = ${batt_capex_target}/kWh, adoption = {adoption}, participation = {selected_participation[0]}",
        fontsize=18,
        y=0.98,
        color="black",
    )

    plt.tight_layout(rect=[0, 0.12, 1, 0.92])

    if save_path is not None:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)

        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight",
            facecolor="white",
        )

    return df, fig, axes

df_line_delta, fig, axes = plot_line_cost_delta_1x3(
    states=("CA", "NY", "TX"),
    adoption="mid",
    batt_capex_target=250,
    baseline_scenario_label_by_state={
        "CA": "midnight",
        "NY": "midnight",
        "TX": "arrive",
    },
    program_scenario_label_by_state={
        "CA": "flex",
        "NY": "flex",
        "TX": "midnight",
    },
    graph_json_path_by_state={
        "CA": "Examples/California_new.json",
        "NY": "Examples/New_York.json",
        "TX": "Examples/Texas.json",
    },
    selected_groups=("V1G", "V2G", "V1G+V2G"),
    selected_participation=("50%",),
    rps_order_by_state={
        "CA": [50, 60, 70],
        "NY": [60, 70, 80],
        "TX": [50, 60, 70],
    },
    amortization=7 / 365,
    save_path="figures/line_cost_delta_1x3.png",
)
```
